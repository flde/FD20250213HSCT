# Network centrality score 

```r
netAnalysis_signalingRole_network <- function(
    
    object, 
    signaling, 
    slot.name="netP", 
    measure=c("outdeg","indeg","flowbet","info"), 
    measure.name=c("Sender","Receiver","Mediator","Influencer"),                          
    color.use=NULL,                      
    font.size=8, 
    font.size.title=10, 
    cluster.rows=FALSE, 
    cluster.cols=FALSE, 
    col_annotation=TRUE, 
    col_split=NULL

) {
    
    # Get matrix
    if (length(slot(object, slot.name)$centr) == 0) {stop("Please run `netAnalysis_computeCentrality` to compute the network centrality scores! ")}
    
    centr <- slot(object, slot.name)$centr[signaling]
    
    for(i in 1:length(centr)) {
        
        centr0 <- centr[[i]]
        mat <- matrix(unlist(centr0), ncol=length(centr0), byrow=FALSE)
        mat <- t(mat)
        rownames(mat) <- names(centr0); colnames(mat) <- names(centr0$outdeg)
        
        if (!is.null(measure)) {
            
            mat <- mat[measure,]
            
            if (!is.null(measure.name)) {
                
                rownames(mat) <- measure.name
                
        }
    }
    
    mat <- sweep(mat, 1L, apply(mat, 1, max), '/', check.margin=FALSE)
    
    # Matrix color
    color.heatmap.use=viridis::mako(100)
        
    mat_color <- c(rev(RColorBrewer::brewer.pal(11,"RdBu"))[1:5], "#ffffff", rev(RColorBrewer::brewer.pal(11,"RdBu"))[7:11])
    mat_breaks <- seq(-max(abs(mat)), max(abs(mat)), length.out=length(mat_color))
    color.heatmap.use <- circlize::colorRamp2(mat_breaks, mat_color) 
    
    # Column annotation 
    df <- data.frame(group=colnames(mat)); rownames(df) <- colnames(mat)
        
    if (is.null(color.use)) {
        
        color.use <- scPalette(length(colnames(mat)))
        cell.cols.assigned <- setNames(color.use, unique(as.character(df$group)))
    
    } else {
        
        cell.cols.assigned <- color.use
        
    }
    
    if(col_annotation) {
        
        col_annotation <- HeatmapAnnotation(
        
            df=df, 
            col=list(group=cell.cols.assigned), 
            which="column", 
            show_legend=FALSE, 
            show_annotation_name=FALSE, 
            simple_anno_size=grid::unit(0.5, "cm")

        )
        
    } else {
        
        col_annotation <- NULL
        
    }
    
    # Set column split 
    if(!is.null(col_split)) {
        
        rownames(col_split) <- col_split$group
        col_split <- col_split[colnames(mat), ]$group_split
        
    }


    hm <- Heatmap(
        
        matrix=mat, 
        name="Importance",
        
        col=color.heatmap.use,
        
        width=unit(ncol(mat), "cm"), 
        height=unit(nrow(mat), "cm"),
        
        column_title=paste0(names(centr[i]), " signaling pathway network"),
        column_title_gp=gpar(fontsize=font.size.title, fontface="bold"),
        
        row_names_gp=gpar(fontsize=font.size),
        column_names_gp=gpar(fontsize=font.size),
        
        row_labels=parse(text=rownames(mat)), 
        column_labels=parse(text=colnames(mat)), 
        
        row_split=NULL, 
        column_split=col_split, 
        
        border=TRUE,
        
        cluster_rows=cluster.rows,
        cluster_columns=cluster.rows,
        
        show_row_names=TRUE,
        show_column_names=TRUE,
        
        row_gap=unit(2, "mm"), 
        column_gap=unit(2, "mm"),
        
        na_col="white", 
        
        bottom_annotation=col_annotation,

        row_names_side="left",
        row_names_rot=0,

        column_names_rot=45,
        
        heatmap_legend_param=list(
            
            title="Importance", 
            title_gp=gpar(fontsize=18, fontface="plain"),
            title_position="leftcenter-rot", 
            border=NA, 
            at=c(round(min(mat, na.rm=T), digits=1), round(max(mat, na.rm=TRUE), digits=1)),                               
            legend_height=unit(5*6, "mm"),
            labels_gp=gpar(fontsize=16),
            grid_width=unit(6, "mm")
        
        )
        
    )
        
    return(hm)
    
    }
    
}
```

## Skin Baseline 

```r
options(repr.plot.width=15, repr.plot.height=5)

hm_1 <- netAnalysis_signalingRole_network(cellchat_1, signaling="MIF", font.size=18, font.size.title=18, color.use=color$cell_type_leiden_subset_r, col_split=cellchat_1@meta[, c("cell_type_leiden_subset", "domain_leiden_subset")] %>% unique() %>% dplyr::rename_with(~c("group", "group_split")))
hm_1
```

```r
png("result/cellchat/figure/mif_signaling_pathway_network_skin_baseline.png", res=300, width=5000, height=2000)
draw(hm_1)
dev.off()
```
