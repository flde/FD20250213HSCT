# Loading and preparing data 

```python
import scvi
import scanpy as sc

import pandas as pd
import subprocess
import os
```

```python
print(scvi.__version__)
```

```python
sc.set_figure_params(figsize=(5, 5))
```

```python
os.chdir('/research/peer/fdeckert/FD20220623HSCT')
```

```python
scvi.settings.num_threads=48
```

# Import AnnData

```python
adata = sc.read_h5ad('data/object/qc.h5ad')
```

# Register model

```python
scvi.model.SCVI.setup_anndata(
    adata,
    categorical_covariate_keys=['patient_id', 'time_point', 'digestion_protocol'], 
    continuous_covariate_keys=['pMT_RNA', 'msCC_diff_RNA']
)
```

# Creating and training a model 

```python
model = scvi.model.SCVI(
    adata, 
    n_latent=10, 
    n_hidden=128, 
    n_layers=2, 
    gene_likelihood='nb'
)
```

```python
model.train()
```

```python
model.save('data/object/scvi/model/', overwrite=True)
```

# Update anndata object with scvi results 

```python
# DEV 
model = scvi.model.SCVI.load('data/object/scvi/model/', adata)
```

```python
adata.layers['data'] = model.get_normalized_expression(library_size=10e4)
adata.obsm['latent'] = model.get_latent_representation()
```

# Dimensional reduction on scvi latent space 

```python
sc.pp.neighbors(adata, n_neighbors=30, use_rep='latent')
sc.tl.leiden(adata, resolution=1, key_added='leiden')
sc.tl.umap(adata, min_dist=1)
```

# Save scvi output adata

```python
# adata.write('data/object/scvi/adata.h5ad')
```

## PAGA and UMAP 

```python
sc.pl.umap(adata, color=['leiden', 'patient_id', 'time_point', 'hto_demux_class', 'digestion_protocol', 'cc_phase_class', 'pMT_RNA', 'pHB_RNA', 'pRP_RNA', 'nCount_RNA', 'nFeature_RNA'], frameon=False, ncols=5, wspace=0.5)
```

## UMAP CellTypist labels with frequency greater 0.1 %

```python
adata.obs['celltypist_predicted_labels_low_hfreq'] = adata.obs['celltypist_predicted_labels_low'].to_list()
adata.obs = adata.obs.apply(lambda x: x.mask(x.map(x.value_counts())<int(adata.n_obs/1000), 'freq < 0.1 %') if x.name=='celltypist_predicted_labels_low_hfreq' else x)

adata.obs['celltypist_predicted_labels_high_hfreq'] = adata.obs['celltypist_predicted_labels_high'].to_list()
adata.obs = adata.obs.apply(lambda x: x.mask(x.map(x.value_counts())<int(adata.n_obs/1000), 'freq < 0.1 %') if x.name=='celltypist_predicted_labels_high_hfreq' else x)
```

```python
sc.pl.umap(adata, color=['leiden', 'celltypist_predicted_labels_low_hfreq', 'celltypist_predicted_labels_high_hfreq'], frameon=False, ncols=1, wspace=0.5)
```
