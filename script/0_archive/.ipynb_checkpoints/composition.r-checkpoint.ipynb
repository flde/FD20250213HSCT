## Pre-process counts and meta data

```r
# Meta data
meta <- dplyr::filter(so@meta.data, time_point %in% c("Baseline", "D14", "D100", "GVHD")) %>% 
    dplyr::mutate(sample_group=paste(sample_name, hto_demux_class)) %>% 
    dplyr::filter(sample_group!="Patient_5_D14 Blood" & sample_group!="Patient_4_D14 Blood" & sample_group!="Patient_1_D14 Blood" & sample_group!="Patient_1_D14 Skin") %>%
    dplyr::select(sample_group, sample_name, gvhd_class, hto_demux_class, digestion_protocol, time_point) %>% 
    unique() 
```

```r
# Cell count data 
counts <- dplyr::mutate(so@meta.data, sample_group=paste(sample_name, hto_demux_class)) %>%
    dplyr::group_by(sample_group, cell_type_leiden_subset) %>% 
    dplyr::summarise(count=n()) %>% 
    dplyr::group_by(sample_group) %>% 
    dplyr::mutate(ratio=count/sum(count)) %>% 
    dplyr::mutate(dropout=ifelse(ratio==0, TRUE, FALSE)) %>% 
    dplyr::ungroup()
```

```r
# Filter mat count by samples from meta 
counts <- counts[counts$sample_group %in% meta$sample_group, ]
```

```r
# Transform mat count to sample group x cell type format and add meta data
counts <- reshape2::dcast(counts, cell_type_leiden_subset~sample_group, value.var="count") %>% tibble::column_to_rownames("cell_type_leiden_subset") %>% replace(is.na(.), 0) 
```

```r
# Get blood and skin cells for subsetting 
blood_cells <- dplyr::filter(so@meta.data, hto_demux_class=="Blood") %>% dplyr::pull(cell_type_leiden_subset) %>% unique() %>% as.character()
skin_cells <- dplyr::filter(so@meta.data, hto_demux_class=="Skin") %>% dplyr::pull(cell_type_leiden_subset) %>% unique() %>% as.character()
```

## Prepare count data 

```r
# Combine count matrix with meta data
mat_count <- counts %>% 
    t() %>% as.data.frame() %>% tibble::rownames_to_column(., "sample_group") %>% 
    dplyr::left_join(., meta, by="sample_group")
```

```r
# Blood data 
mat_count_1 <- mat_count[mat_count["time_point"]=="Baseline" & mat_count["hto_demux_class"]=="Blood", colnames(mat_count) %in% c(blood_cells, colnames(meta))]
mat_count_2 <- mat_count[mat_count["time_point"]=="D14" & mat_count["hto_demux_class"]=="Blood", colnames(mat_count) %in% c(blood_cells, colnames(meta))]
mat_count_3 <- mat_count[mat_count["time_point"]=="D100" & mat_count["hto_demux_class"]=="Blood", colnames(mat_count) %in% c(blood_cells, colnames(meta))]
mat_count_4 <- mat_count[((mat_count["time_point"]=="D100" & mat_count["gvhd_class"]=="GVHD_free") | mat_count["time_point"]=="GVHD") & mat_count["hto_demux_class"]=="Blood", colnames(mat_count) %in% c(blood_cells, colnames(meta))]

# Skin time points 
mat_count_5 <- mat_count[mat_count["time_point"]=="Baseline" & mat_count["hto_demux_class"]=="Skin", colnames(mat_count) %in% c(skin_cells, colnames(meta))] 
mat_count_6 <- mat_count[mat_count["time_point"]=="D14" & mat_count["hto_demux_class"]=="Skin", colnames(mat_count) %in% c(skin_cells, colnames(meta))]
mat_count_7 <- mat_count[mat_count["time_point"]=="D100" & mat_count["hto_demux_class"]=="Skin", colnames(mat_count) %in% c(skin_cells, colnames(meta))]
mat_count_8 <- mat_count[((mat_count["time_point"]=="D100" & mat_count["gvhd_class"]=="GVHD_free") | mat_count["time_point"]=="GVHD") & mat_count["hto_demux_class"]=="Skin", colnames(mat_count) %in% c(skin_cells, colnames(meta))]
```

## Prepare CPM data 

```r
# CPM with library normalizing factors 
mat_cpm_1 <- edgeR::DGEList(counts=counts[, grepl("Blood", colnames(counts))])
mat_cpm_1 <- edgeR::calcNormFactors(mat_cpm_1)
mat_cpm_1 <- edgeR::cpm(mat_cpm_1, log=FALSE, normalized.lib.sizes=TRUE, prior.count=0)

mat_cpm_2 <- edgeR::DGEList(counts=counts[, grepl("Skin", colnames(counts))])
mat_cpm_2 <- edgeR::calcNormFactors(mat_cpm_2)
mat_cpm_2 <- edgeR::cpm(mat_cpm_2, log=FALSE, normalized.lib.sizes=TRUE, prior.count=0)

mat_cpm <- cbind(mat_cpm_1, mat_cpm_2)
```

```r
# Combine count matrix with meta data
mat_cpm <- mat_cpm %>% 
    t() %>% as.data.frame() %>% tibble::rownames_to_column(., "sample_group") %>% 
    dplyr::left_join(., meta, by="sample_group")
```

```r
# Blood data 
mat_cpm_1 <- mat_cpm[mat_cpm["time_point"]=="Baseline" & mat_cpm["hto_demux_class"]=="Blood", colnames(mat_cpm) %in% c(blood_cells, colnames(meta))]
mat_cpm_2 <- mat_cpm[mat_cpm["time_point"]=="D14" & mat_cpm["hto_demux_class"]=="Blood", colnames(mat_cpm) %in% c(blood_cells, colnames(meta))]
mat_cpm_3 <- mat_cpm[mat_cpm["time_point"]=="D100" & mat_cpm["hto_demux_class"]=="Blood", colnames(mat_cpm) %in% c(blood_cells, colnames(meta))]
mat_cpm_4 <- mat_cpm[((mat_cpm["time_point"]=="D100" & mat_cpm["gvhd_class"]=="GVHD_free") | mat_cpm["time_point"]=="GVHD") & mat_cpm["hto_demux_class"]=="Blood", colnames(mat_cpm) %in% c(blood_cells, colnames(meta))]

# Skin time points 
mat_cpm_5 <- mat_cpm[mat_cpm["time_point"]=="Baseline" & mat_cpm["hto_demux_class"]=="Skin", colnames(mat_cpm) %in% c(skin_cells, colnames(meta))] 
mat_cpm_6 <- mat_cpm[mat_cpm["time_point"]=="D14" & mat_cpm["hto_demux_class"]=="Skin", colnames(mat_cpm) %in% c(skin_cells, colnames(meta))]
mat_cpm_7 <- mat_cpm[mat_cpm["time_point"]=="D100" & mat_cpm["hto_demux_class"]=="Skin", colnames(mat_cpm) %in% c(skin_cells, colnames(meta))]
mat_cpm_8 <- mat_cpm[((mat_cpm["time_point"]=="D100" & mat_cpm["gvhd_class"]=="GVHD_free") | mat_cpm["time_point"]=="GVHD") & mat_cpm["hto_demux_class"]=="Skin", colnames(mat_cpm) %in% c(skin_cells, colnames(meta))]
```

## Helper function 

```r
sccoda_cycler <- function(data, formula, out_dir) {
    
    cell_types <- data$var_names$to_list()
    results_cycle <- list()

    for (i in 1:length(cell_types)) {
        
        # Run inference
        model_temp <- mod$CompositionalAnalysis(data, formula=formula, reference_cell_type=cell_types[i])
        tmp_result <- model_temp$sample_hmc()
        
        # Save results 
        tmp_result$save(paste0(out_dir, "cycle_", i, ".pkl"))

    }
    
}
```

```r
sccoda_cycler_import <- function(data, est_fdr, out_dir, restrict=FALSE) {
    
    cell_types <- data$var_names$to_list()
    out_files <- list.files(out_dir, full.names=TRUE)
    
    import <- list()
    
    for (i in 1:length(out_files)) {
        
        tmp_result <- py_load_object(out_files[i], pickle="pickle", convert=TRUE)
        tmp_result$set_fdr(est_fdr=est_fdr)
        effects <- tmp_result$effect_df
        effects$effects_credible <- tmp_result$credible_effects()
        effects$cell_type <- cell_types
        effects$reference <- cell_types[i]
        
        if(restrict) {effects <- effects[17:56, ]}
        
        import[[i]] <- effects
        
    }
    
    import <- do.call(rbind, import)
    
    return(import)
    
}
```

```r
sccoda_cycler_credibility <- function(cycler_import, data, pct_credible_thr) {
    
    cell_types <- data$var_names$to_list()
    
    cycler_credibility <- dplyr::group_by(cycler_import, cell_type) %>% dplyr::summarise(times_credible=sum(effects_credible)) %>% dplyr::mutate(pct_credible=times_credible/length(cell_types)) %>% dplyr::mutate(is_credible=pct_credible>pct_credible_thr)
    times_log2fc <- dplyr::group_by(cycler_import, cell_type) %>% dplyr::filter(effects_credible) %>% dplyr::summarise(times_log2fc=abs(sum(sign(`log2-fold change`))))
    mean_log2fc <- dplyr::group_by(cycler_import, cell_type) %>% dplyr::filter(effects_credible) %>% dplyr::summarise(mean_log2fc=mean(`log2-fold change`))
    
    cycler_credibility <- dplyr::left_join(cycler_credibility, times_log2fc, by=join_by(cell_type)) %>% replace(is.na(.), 0)
    cycler_credibility <- dplyr::left_join(cycler_credibility, mean_log2fc, by=join_by(cell_type)) %>% dplyr::mutate(mean_log2fc=ifelse(is_credible, mean_log2fc, 0))
    
    cycler_credibility <- dplyr::rename(cycler_credibility, cell_type_leiden_subset="cell_type")
    
    return(cycler_credibility)
    
}
```

```r
shannon <- data.frame(sample_group=colnames(cell_count_cnt), shannon=vegan::diversity(t(cell_count_cnt), index="simpson", base=2)) %>% dplyr::left_join(., dplyr::select(cell_count, sample_group, family_domain_subset, hto_demux_class, patient_id, time_point) %>% unique(), by="sample_group")
```

## Time points by sample and genotype 

```r
cell_count <- dplyr::mutate(so@meta.data, sample_group=paste(sample_name, genotype_class, family_domain_subset, hto_demux_class)) %>%
    dplyr::group_by(sample_group, sample_name, genotype_class, family_domain_subset, hto_demux_class, time_point, cell_type_leiden_subset) %>%  
    dplyr::summarise(cell_count=n())
```

```r
cell_count_cnt <- dcast(cell_count, cell_type_leiden_subset~sample_group, value.var="cell_count") %>% tibble::column_to_rownames("cell_type_leiden_subset") %>% replace(is.na(.), 0)
```

```r
shannon <- data.frame(sample_group=colnames(cell_count_cnt), shannon=vegan::diversity(t(cell_count_cnt), index="simpson", base=2)) %>% 
    dplyr::left_join(., dplyr::select(cell_count, sample_group, sample_name, genotype_class, family_domain_subset, hto_demux_class, time_point) %>% unique(), by="sample_group") %>% 
    dplyr::group_by(genotype_class, family_domain_subset, hto_demux_class, time_point) %>% 
    dplyr::summarise(shannon_median=median(shannon), shannon_mad=mad(shannon, constant=1)) %>%
    dplyr::ungroup() %>% 
    tidyr::complete(genotype_class, family_domain_subset, hto_demux_class, time_point, fill=list(shannon_median=0, shannon_mad=0)) %>% 
    dplyr::filter(genotype_class %in% c("Host", "Donor"))
```

```r
options(repr.plot.width=10, repr.plot.height=3)

dodge_with <- 0.5
mad_scale <- 2

plot_4 <- ggplot(shannon[shannon$hto_demux_class=="Blood" & shannon$family_domain_subset=="Immune cell", ], aes(x=time_point, y=shannon_median, color=genotype_class)) + 
    geom_line(aes(group=genotype_class), size=1, position=position_dodge(dodge_with)) + 
    geom_point(shape=16, size=3, position=position_dodge(dodge_with)) + 
    geom_errorbar(aes(ymin=ifelse(shannon_median-mad_scale*shannon_mad<0, 0, shannon_median-mad_scale*shannon_mad), ymax=shannon_median+mad_scale*shannon_mad), width=0.5, size=0.75, position=position_dodge(dodge_with)) + 
    ggtitle("Blood") + xlab("") + ylab("Shannon") +
    ylim(0, 1) +
    scale_color_manual(values=color$genotype_class, name="Genotype class") + 
    theme(axis.text.x=element_text(angle=90, vjust=0.5, hjust=1))

plot_5 <- ggplot(shannon[shannon$hto_demux_class=="Skin" & shannon$family_domain_subset=="Immune cell", ], aes(x=time_point, y=shannon_median, color=genotype_class)) + 
    geom_line(aes(group=genotype_class), size=1, position=position_dodge(dodge_with)) + 
    geom_point(shape=16, size=3, position=position_dodge(dodge_with)) + 
    geom_errorbar(aes(ymin=ifelse(shannon_median-mad_scale*shannon_mad<0, 0, shannon_median-mad_scale*shannon_mad), ymax=shannon_median+mad_scale*shannon_mad), width=0.5, size=0.75, position=position_dodge(dodge_with)) + 
    ggtitle("Skin (Immune)") + xlab("") + ylab("Shannon") +
    ylim(0, 1) +
    scale_color_manual(values=color$genotype_class, name="Genotype class") + 
    theme(axis.text.x=element_text(angle=90, vjust=0.5, hjust=1))

plot_6 <- ggplot(shannon[shannon$hto_demux_class=="Skin" & shannon$family_domain_subset=="Non immune cell", ], aes(x=time_point, y=shannon_median, color=genotype_class)) + 
    geom_line(aes(group=genotype_class), size=1, position=position_dodge(dodge_with)) + 
    geom_point(shape=16, size=3, position=position_dodge(dodge_with)) + 
    geom_errorbar(aes(ymin=ifelse(shannon_median-mad_scale*shannon_mad<0, 0, shannon_median-mad_scale*shannon_mad), ymax=shannon_median+mad_scale*shannon_mad), width=0.5, size=0.75, position=position_dodge(dodge_with)) + 
    ggtitle("Skin (Non immune)") + xlab("") + ylab("Shannon") +
    ylim(0, 1) +
    scale_color_manual(values=color$genotype_class, name="Genotype class") + 
    theme(axis.text.x=element_text(angle=90, vjust=0.5, hjust=1))
```

## Genotype ratio by time points 

```r
genotype_ratio <- dplyr::group_by(so@meta.data, family_domain_subset, patient_id, hto_demux_class, time_point, genotype_class) %>%  
    dplyr::summarise(cell_count=n()) %>% dplyr::ungroup() %>% 
    tidyr::complete(family_domain_subset, patient_id, hto_demux_class, time_point, genotype_class, fill=list(cell_count=0)) %>% 
    dplyr::group_by(family_domain_subset, patient_id, hto_demux_class, time_point) %>% 
    dplyr::mutate(library_size=sum(cell_count)) %>% dplyr::ungroup() %>% 
    dplyr::filter(genotype_class %in% c("Host", "Donor")) %>% 
    dplyr::mutate(ratio=cell_count/library_size) %>% 
    dplyr::group_by(family_domain_subset, hto_demux_class, time_point, genotype_class) %>% 
    dplyr::summarise(ratio_mean=mean(ratio, na.rm=TRUE), ratio_sd=sd(ratio, na.rm=TRUE))
```

```r
options(repr.plot.width=10, repr.plot.height=3)

dodge_with <- 0.5

plot_7 <- ggplot(genotype_ratio[genotype_ratio$hto_demux_class=="Blood" & genotype_ratio$family_domain_subset=="Immune cell", ], aes(x=time_point, y=ratio_mean, color=genotype_class)) + 
    geom_line(aes(group=genotype_class), size=1, position=position_dodge(dodge_with)) + 
    geom_point(shape=16, size=3, position=position_dodge(dodge_with)) + 
    geom_errorbar(aes(ymin=ifelse(ratio_mean-ratio_sd<0, 0, ratio_mean-ratio_sd), ymax=ifelse(ratio_mean+ratio_sd>1, 1, ratio_mean+ratio_sd)), width=0.5, size=0.75, position=position_dodge(dodge_with)) + 
    ggtitle("Blood") + xlab("") + ylab("Ratio") +
    ylim(0, 1) +
    scale_color_manual(values=color$genotype_class, name="Genotype class") + 
    theme(axis.text.x=element_text(angle=90, vjust=0.5, hjust=1))

plot_8 <- ggplot(genotype_ratio[genotype_ratio$hto_demux_class=="Skin" & genotype_ratio$family_domain_subset=="Immune cell", ], aes(x=time_point, y=ratio_mean, color=genotype_class)) + 
    geom_line(aes(group=genotype_class), size=1, position=position_dodge(dodge_with)) + 
    geom_point(shape=16, size=3, position=position_dodge(dodge_with)) + 
    geom_errorbar(aes(ymin=ifelse(ratio_mean-ratio_sd<0, 0, ratio_mean-ratio_sd), ymax=ifelse(ratio_mean+ratio_sd>1, 1, ratio_mean+ratio_sd)), width=0.5, size=0.75, position=position_dodge(dodge_with)) + 
    ggtitle("Skin (Immune)") + xlab("") + ylab("Ratio") +
    ylim(0, 1) +
    scale_color_manual(values=color$genotype_class, name="Genotype class") + 
    theme(axis.text.x=element_text(angle=90, vjust=0.5, hjust=1))

plot_9 <- ggplot(genotype_ratio[genotype_ratio$hto_demux_class=="Skin" & genotype_ratio$family_domain_subset=="Non immune cell", ], aes(x=time_point, y=ratio_mean, color=genotype_class)) + 
    geom_line(aes(group=genotype_class), size=1, position=position_dodge(dodge_with)) + 
    geom_point(shape=16, size=3, position=position_dodge(dodge_with)) + 
    geom_errorbar(aes(ymin=ifelse(ratio_mean-ratio_sd<0, 0, ratio_mean-ratio_sd), ymax=ifelse(ratio_mean+ratio_sd>1, 1, ratio_mean+ratio_sd)), width=0.5, size=0.75, position=position_dodge(dodge_with)) + 
    ggtitle("Skin (Non immune)") + xlab("") + ylab("Ratio") +
    ylim(0, 1) +
    scale_color_manual(values=color$genotype_class, name="Genotype class") + 
    theme(axis.text.x=element_text(angle=90, vjust=0.5, hjust=1))
```
