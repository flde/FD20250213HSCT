# GT variants import

```r
gt_import <- function(patient_id, gtf) {
    
    # Import GT donor file
    gt_vcf <- read.table(paste0("data/object/genotype/vireo/vireo_mode_1a_genome1k/", patient_id, "_vireo_cellsnp_0/GT_donors.vireo.vcf.gz"), header=FALSE)
    colnames(gt_vcf) <- c("CHROM", "POS", "ID", "REF", "ALT", "QUAL", "FILTER", "INFO", "FORMAT", "DONOR0", "DONOR1")
    gt_vcf$ID <- paste0(gt_vcf$CHROM, "_", gt_vcf$POS, "_", gt_vcf$REF, "_", gt_vcf$ALT)
    
    # Import variant tsv
    variant_vcf_file <- paste0("data/object/genotype/variants/vireo_mode_1a_genome1k/", patient_id, "_vireo_cellsnp_0/variants.tsv")
    variant_vcf <- read.table(variant_vcf_file, header=TRUE)
    colnames(variant_vcf) <- c("ID", "DONOR0", "DONOR1")
    
    # Filter GT donor by variants 
    gt_vcf$VARIANT <- gt_vcf$ID %in% variant_vcf$ID
    
    # Set variables 
    gt_vcf$GT_DONOR0 <- sapply(strsplit(gt_vcf$DONOR0, ":"), "[", 1) 
    gt_vcf$GT_DONOR1 <- sapply(strsplit(gt_vcf$DONOR1, ":"), "[", 1)
    gt_vcf$AD_DONOR0 <- as.numeric(sapply(strsplit(gt_vcf$DONOR0, ":"), "[", 2))
    gt_vcf$AD_DONOR1 <- as.numeric(sapply(strsplit(gt_vcf$DONOR1, ":"), "[", 2))
    gt_vcf$DP_DONOR0 <- as.numeric(sapply(strsplit(gt_vcf$DONOR0, ":"), "[", 3))
    gt_vcf$DP_DONOR1 <- as.numeric(sapply(strsplit(gt_vcf$DONOR1, ":"), "[", 3))
    
    gt_vcf$AD <- sapply(strsplit(sapply(strsplit(gt_vcf$INFO, ";"), "[", 1), "="), "[", 2) %>% as.numeric() 
    gt_vcf$DP <- sapply(strsplit(sapply(strsplit(gt_vcf$INFO, ";"), "[", 2), "="), "[", 2) %>% as.numeric() 
    gt_vcf$OTH <- sapply(strsplit(sapply(strsplit(gt_vcf$INFO, ";"), "[", 3), "="), "[", 2) %>% as.numeric()
    gt_vcf$CNT <- sum(gt_vcf$AD) + sum(gt_vcf$DP) + sum(gt_vcf$OTH)

    gt_vcf$PATIENT_ID <- patient_id
    
    return(gt_vcf)
    
}
```

```r
gt_vcf <- lapply(paste0("Patient_", 1:11), gt_import, gtf=gtf)
gt_vcf <- do.call("rbind", gt_vcf)
```

```r
gt_vcf$PATIENT_ID <- factor(gt_vcf$PATIENT_ID, levels=paste0("Patient_", 1:11))
```

```r
dplyr::filter(gt_vcf, DP_DONOR0>10, DP_DONOR1>10) %>% 
    dplyr::select(ID, PATIENT_ID, CNT, VARIANT) %>% distinct() %>% 
    group_by(PATIENT_ID) %>% dplyr::mutate(N_VARIANTS=n()) %>% dplyr::select(-ID) %>% 
    distinct() %>% dplyr::mutate(NORM_VARIANTS=10^6*N_VARIANTS/CNT) %>% 
    dplyr::mutate(PATIENT_ID=factor(PATIENT_ID, levels=paste0("Patient_", 1:11))) %>% 
    ggplot(., aes(x=PATIENT_ID, y=NORM_VARIANTS, fill=VARIANT)) + geom_bar(stat="identity") + theme(axis.text.x = element_text(angle=90, vjust=0.5, hjust=1))
```

```r
vep <- read.table(paste0("data/object/genotype/vep/", "Patient_1", "/variant_effect_output.txt"))
```

```r
vep_import <- function(PATIENT_ID) {
    
    vep <- read.table(paste0("data/object/genotype/vep/", PATIENT_ID, "/variant_effect_output.txt"))
    colnames(vep) <- toupper(c("id", "Location", "Allele", "Gene", "Feature", "Feature_type", "Consequence", "cDNA_position", "CDS_position", "Protein_position", "Amino_acids", "Codons", "Existing_variation", "Extra"))
    
    # Annotate 
    ref <- as.data.frame(gtf)[, c("gene_id", "gene_name")] %>% na.omit() %>% unique()
    colnames(ref) <- c("GENE", "SYMBOL")
    
    vep <- dplyr::left_join(vep, ref, by = join_by(GENE))
    
    # Fix variant ID
    vep$ID <- gsub("/", "_", vep$ID)
    
    vep <- vep[, c("ID", "ALLELE", "FEATURE_TYPE", "CONSEQUENCE", "SYMBOL")]
    
    # Set patient id
    vep$PATIENT_ID <- PATIENT_ID
    
    return(vep)
}
```

```r
vep <- lapply(paste0("Patient_", 1:11), vep_import)
vep <- do.call("rbind", vep)
```

# Plot missense variants by SNP ID

```r
mat <- dplyr::filter(data, OTH<AD, GT_DONOR0!=GT_DONOR1, CONSEQUENCE=="missense_variant") %>% distinct %>% 
    dplyr::group_by(ID, SYMBOL) %>% 
    dplyr::mutate(ID_COUNT_GVHD_FREE=sum(GVHD_CLASS=="GVHD_free"), ID_COUNT_GVHD_DEV=sum(GVHD_CLASS=="GVHD_dev")) %>% 
    dplyr::arrange(-ID_COUNT_GVHD_DEV, ID_COUNT_GVHD_FREE, ID) %>% dplyr::ungroup() %>% 
    dplyr::mutate(ID=paste0(SYMBOL, " (", ID, ""))
```

```r
row_annotation_df <- dplyr::select(mat, PATIENT_ID, GVHD_CLASS) %>% distinct %>% arrange(PATIENT_ID) 
rownames(row_annotation_df) <- row_annotation_df$PATIENT_ID
```

```r
row_annotation <- rowAnnotation(

    GVHD_CLASS=row_annotation_df[, "GVHD_CLASS", drop=TRUE], 

    annotation_legend_param=list(

        GVHD_CLASS=list(title="GVHD status", title_gp=gpar(fontsize=18, fontface="plain"), labels_gp=gpar(fontsize=16), grid_width=unit(6, "mm"), grid_height=unit(6, "mm"))

    ), 

    col=list(

        GVHD_CLASS=setNames(c("#5A5A5A", "#B3254D"), c("GVHD_free", "GVHD_dev"))

    ), 

    simple_anno_size=unit(5.5, "mm"), 
    show_annotation_name=FALSE, 
    show_legend=TRUE

)    
```

```r
VARIANT_ID_GVHD_FREE <- dplyr::filter(mat, VARIANT, GVHD_CLASS=="GVHD_free", ID_COUNT_GVHD_FREE>=3) %>% dplyr::pull(ID) %>% unique()
VARIANT_ID_GVHD_DEV <- dplyr::filter(mat, VARIANT, GVHD_CLASS=="GVHD_dev", ID_COUNT_GVHD_DEV>=5) %>% dplyr::pull(ID) %>% unique()
```

```r
VARIANT_ID_SHARED <- intersect(VARIANT_ID_GVHD_FREE, VARIANT_ID_GVHD_DEV)
VARIANT_ID_GVHD_FREE <- VARIANT_ID_GVHD_FREE[!VARIANT_ID_GVHD_FREE %in% VARIANT_ID_SHARED]
VARIANT_ID_GVHD_DEV <- VARIANT_ID_GVHD_DEV[!VARIANT_ID_GVHD_DEV %in% VARIANT_ID_SHARED]
```

```r
column_split <- c(rep("SHARED", length(VARIANT_ID_SHARED)), rep("FREE", length(VARIANT_ID_GVHD_FREE)), rep("DEV", length(VARIANT_ID_GVHD_DEV)))
```

```r
mat <- dplyr::mutate(data, ID=paste0(SYMBOL, " (", ID, "")) %>% dplyr::filter(ID %in% c(VARIANT_ID_SHARED, VARIANT_ID_GVHD_FREE, VARIANT_ID_GVHD_DEV)) %>% dplyr::mutate(MAF=ifelse(DP>20 & AD>10, AD/DP, 0)) %>% dplyr::select(PATIENT_ID, ID, MAF) %>% distinct()
mat <- reshape2::dcast(mat, PATIENT_ID~ID, value.var="MAF", fill=0) %>% tibble::column_to_rownames("PATIENT_ID")
mat <- mat[, c(VARIANT_ID_SHARED, VARIANT_ID_GVHD_FREE, VARIANT_ID_GVHD_DEV)]
```

```r
# Color function 
color_ramp_mat <- viridis::inferno(10)
breaks_mat <- seq(0, max(mat), length.out=10)
color_function_mat <- circlize::colorRamp2(breaks_mat, color_ramp_mat) 
```

```r
options(repr.plot.width=40, repr.plot.height=10)

hm <- ComplexHeatmap::Heatmap(
    
    matrix=mat, 

    col=color_function_mat, 
    na_col="white", 

    width=unit(6*ncol(mat), "mm"), 
    height=unit(6*nrow(mat), "mm"), 

    row_title=NULL, 
    row_title_gp=gpar(fontsize=18, fontface="bold"), 

    # column_title="DEA gene module", 
    column_title_gp=gpar(fontsize=18, fontface="bold"), 

    row_names_gp=gpar(fontsize=16, fontface="plain"), 
    column_names_gp=gpar(fontsize=18, fontface="plain"), 

    # cluster_rows=function(mat) hclust(dist(mat), method="complete"), 
    cluster_rows=TRUE, 
    show_row_dend=TRUE,   
    row_split=row_annotation_df$GVHD_CLASS, 
    row_gap=unit(2, "mm"),
    show_row_names=TRUE,
    row_names_side="right",

    cluster_columns=TRUE,
    cluster_column_slices=TRUE, 
    show_column_dend=TRUE, 
    column_split=column_split,
    column_gap=unit(2, "mm"), 
    show_column_names=TRUE, 

    top_annotation=NULL, 
    left_annotation=row_annotation, 

    rect_gp=gpar(col=NA, lwd=0, alpha=1), 

    heatmap_legend_param=list(title="MAF", title_gp=gpar(fontsize=18, fontface="plain"), labels_gp=gpar(fontsize=16), legend_height=unit(5*6, "mm"), grid_width=unit(6, "mm")), 

    border=TRUE, 
    border_gp=gpar(col="black", lwd=unit(0.5, "mm")), 

    use_raster=FALSE

)

hm
```
