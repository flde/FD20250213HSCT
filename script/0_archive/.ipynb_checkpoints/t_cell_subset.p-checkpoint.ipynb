```python
# Load color 
import rpy2.robjects as robjects
color_load = robjects.r.source('plotting_global.R')
color = dict()
for i in range(len(color_load[0])):
    color[color_load[0].names[i]] = {key : color_load[0][i].rx2(key)[0] for key in color_load[0][i].names}

# Set color function 
def set_color(adata, categories): 
    
    categories = [x for x in categories if x in list(adata.obs.columns)]

    for category in categories: 

        adata.obs[category] = pd.Series(adata.obs[category], dtype='category').cat.remove_unused_categories()
        
        keys = list(color[category].keys())
        keys = [x for x in keys if x in list(adata.obs[category])]        

        adata.obs[category] = adata.obs[category].cat.reorder_categories(keys)
        adata.uns[category+'_colors'] = np.array([color[category].get(key) for key in keys], dtype=object)
        
    return(adata)
```

```python
def module_score(adata, marker_genes_dict): 
    
    # Normalize data 
    sc.pp.normalize_total(adata)
    sc.pp.log1p(adata)
    
    for items in marker_genes_dict.items(): 
        sc.tl.score_genes(adata, gene_list=items[1], score_name=items[0], use_raw=False)
        
    module_score_mtx = adata.obs[list(marker_genes_dict.keys())]
        
    return(module_score_mtx)
```

```python
adata.obs = adata.obs.merge(module_score(adata.copy(), marker_genes_pool_dict), left_index=True, right_index=True, how='left')
```

# HVG selection 

```python
def hvg_select(adata, key, key_min_size=100, key_min_n=2): 
    
    # Check key size
    key_size = adata.obs[key].value_counts()
    print('Sample size: '+ str(len(key_size)))
    
    # Select key by size 
    key_size = key_size[key_size>key_min_size]
    key_select = key_size.index.to_list()
    print('Sample size select: '+ str(len(key_size)))
    
    # Subset by selected keys
    adata = adata[adata.obs[key].isin(key_select)]
    
    # Subset genes
    adata = adata[:, (adata.X>0).sum(axis=0)>=1]
    
    # Normalize data 
    sc.pp.normalize_total(adata)
    sc.pp.log1p(adata)
    
    # Highly variable genes 
    sc.pp.highly_variable_genes(adata, flavor='seurat', min_disp=0.25, batch_key=key)
    
    hvg = adata.var_names[(adata.var['highly_variable']) & (adata.var['highly_variable_nbatches']>=key_min_n)].to_list()
    print('Number hvg :' + str(len(hvg)))
    
    return(hvg)
```

```python
hvg_1 = hvg_select(adata[adata.obs['time_point'].isin(['Baseline'])].copy(), key='sample_name', key_min_n=3, key_min_size=500)
hvg_2 = hvg_select(adata[adata.obs['time_point'].isin(['D14'])].copy(), key='sample_name', key_min_n=2, key_min_size=250)
hvg_3 = hvg_select(adata[adata.obs['time_point'].isin(['D100'])].copy(), key='sample_name', key_min_n=2, key_min_size=250)
hvg_4 = hvg_select(adata[adata.obs['time_point'].isin(['D100'])].copy(), key='sample_name', key_min_n=2, key_min_size=250)
hvg_5 = hvg_select(adata[adata.obs['time_point'].isin(['GVHD'])].copy(), key='sample_name', key_min_n=2, key_min_size=250)
```

```python
hvg = pd.Series([*hvg_1, *hvg_2, *hvg_3, *hvg_4, *hvg_5], dtype="category").value_counts()
hvg = hvg[hvg>=1].index.to_list()
len(hvg)
```

# Harmony integratolist 

```python
def scanorama_integrate(adata, hvg, key, dimred=100, file=None):
    
    # Subset genes
    adata = adata[:, (adata.X>3).sum(axis=0)>=3]
    
    # Normalize data 
    sc.pp.normalize_total(adata)
    sc.pp.log1p(adata)
    
    # Scale 
    sc.pp.scale(adata, zero_center=True)
    
    # Subset by hvg 
    adata = adata[:, adata.var_names.isin(hvg)]
    
    # Subset data by i  
    adata_key = dict()
    for i in adata.obs[key].unique():

        adata_key[i] = adata[adata.obs[key]==i]
        
    adata_key = list(adata_key.values())
    
    # Scanorama with count matrix correction 
    scanorama.integrate_scanpy(adata_key, dimred=dimred, verbose=False)
    adata = ad.concat(adata_key, join="inner")
    
    # Set colors
    adata = set_color(adata, list(color.keys()))
    
    # Reset full gene matrix 
    adata_raw = adata.raw.to_adata()
    
    adata = adata.raw.to_adata()
    adata.raw = adata_raw
    
    # wirte output 
    if file is not None: 
        adata.write_h5ad(file)
    
    return(adata)
```

```python
adata_scanorama = scanorama_integrate(adata.copy(), hvg=hvg, key='scanorama_integrate', file=None)
```

# Dim reduction and cluster function 

```python
def dim_clust(adata, n_neighbors=10, resolution=3, min_dist=0.5, use_rep='X_scanorama', metric='euclidean', file=None): 
    
    # Dim reduction and clustering on the scanorama corrected embedding 
    sc.pp.neighbors(adata, n_neighbors=n_neighbors, use_rep=use_rep, metric=metric)
    sc.tl.leiden(adata, resolution=resolution)
    
    sc.tl.umap(adata, min_dist=min_dist)
    
    # wirte output 
    if file is not None: 
        adata.write_h5ad(file)
    
    return(adata)
```
