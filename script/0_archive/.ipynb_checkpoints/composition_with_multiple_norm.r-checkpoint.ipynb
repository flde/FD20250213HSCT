# Cell type distribution 

## Pre-process data

```r
# Meta data with cell counts per sample 
cell_count <- dplyr::group_by(so@meta.data, sample_name, time_point, gvhd_class, digestion_protocol, cell_type_leiden_subset) %>% 
    dplyr::summarise(cell_count=n()) %>% 
    dplyr::mutate(time_point=factor(time_point, levels=names(color$time_point))) %>% 
    dplyr::mutate(cell_type_leiden_subset=factor(cell_type_leiden_subset, levels=names(color$cell_type_leiden_subset_r))) %>% 
    dplyr::arrange(time_point) 

cell_count$sample_name <- factor(cell_count$sample_name, levels=unique(cell_count$sample_name))
```

```r
# Cell count matrix cell type x samples 
cell_count_mtx <- dcast(cell_count, cell_type_leiden_subset~sample_name, value.var="cell_count") %>% tibble::column_to_rownames("cell_type_leiden_subset") %>% replace(is.na(.), 0)
```

```r
# Save 0 count matrix 
mask <- cell_count_mtx==0
```

```r
# Library normalizing factors 
cell_count_norm <- edgeR::DGEList(counts=cell_count_mtx)
cell_count_norm <- edgeR::calcNormFactors(cell_count_norm)
```

```r
# Covariants matrix 
covar <- cell_count[, c("sample_name", "digestion_protocol", "time_point", "gvhd_class")] %>% unique()

# Batch variable by digestion protocol 
batch <- covar$digestion_protocol

# Model matrix 
covar$gvhd_class <- factor(covar$gvhd_class, levels=c("GVHD_free", "GVHD_dev"))
covar_mod <- model.matrix(~time_point+gvhd_class, covar)
rownames(covar_mod) <- covar$sample_name
```

## Counts

```r
options(repr.plot.width=15, repr.plot.height=7.5)

plot_1 <- ggplot(cell_count, aes(x=sample_name, y=cell_count, fill=cell_type_leiden_subset)) + 
    geom_bar(stat="identity") + 
    ggtitle("Counts") + xlab("Sample name") + ylab("Cell type [Counts]")+ 
    scale_fill_manual(values=color$cell_type_leiden_subset_r, name="Cell type", labels=scales::parse_format(),) +
    theme(axis.text.x = element_text(angle=90, vjust=0.5, hjust=1, colour=ifelse(unique(cell_count[, c("sample_name", "digestion_protocol")])$digestion_protocol=="new", "#377EB8", "#E41A1C"))) + 
    guides(fill=guide_legend(ncol=2))

plot_2 <- ggplot(cell_count, aes(x=sample_name, y=cell_count, fill=cell_type_leiden_subset)) + 
    geom_bar(stat="identity", position="fill") + 
    ggtitle("Cell type distribution") + xlab("Sample name") + ylab("Cell type [Relative]")+ 
    scale_fill_manual(values=color$cell_type_leiden_subset_r, name="Cell type", labels=scales::parse_format(),) +
    theme(axis.text.x = element_text(angle=90, vjust=0.5, hjust=1, colour=ifelse(unique(cell_count[, c("sample_name", "digestion_protocol")])$digestion_protocol=="new", "#377EB8", "#E41A1C"))) + 
    guides(fill=guide_legend(ncol=2))
```

# Shannon entropy

## Time points by sample 

```r
cell_count <- dplyr::mutate(so@meta.data, group=paste(sample_name, family_domain_subset, hto_demux_class)) %>%
    dplyr::group_by(group, sample_name, patient_id, family_domain_subset, hto_demux_class, time_point, cell_type_leiden_subset) %>%  
    dplyr::summarise(cell_count=n())

cell_count_cnt <- dcast(cell_count, cell_type_leiden_subset~group, value.var="cell_count") %>% tibble::column_to_rownames("cell_type_leiden_subset") %>% replace(is.na(.), 0)
```

```r
shannon <- data.frame(group=colnames(cell_count_cnt), shannon=vegan::diversity(t(cell_count_cnt), index="simpson", base=2)) %>% dplyr::left_join(., dplyr::select(cell_count, group, family_domain_subset, hto_demux_class, patient_id, time_point) %>% unique(), by="group")
```

```r
options(repr.plot.width=10, repr.plot.height=3)

plot_1 <- ggplot(shannon[shannon$hto_demux_class=="Blood", ], aes(x=time_point, y=shannon)) + 
    geom_boxplot(aes(fill=time_point), outlier.shape=1, outlier.size=3, size=0.5, color="black") + 
    geom_point(size=3, shape=1) + 
    ggtitle("Blood") + xlab("") + ylab("Shannon") +
    ylim(0, max(shannon$shannon)+0.1) +
    scale_fill_manual(values=color$time_point, name="Time point") + 
    theme(axis.text.x=element_blank())

plot_2 <- ggplot(shannon[shannon$hto_demux_class=="Skin" & shannon$family_domain_subset=="Immune cell", ], aes(x=time_point, y=shannon)) + 
    geom_boxplot(aes(fill=time_point), outlier.shape=1, outlier.size=3, size=0.5, color="black") + 
    geom_point(size=3, shape=1) + 
    ggtitle("Skin (Immune)") + xlab("") + ylab("Shannon") +
    ylim(0, max(shannon$shannon)+0.1) +
    scale_fill_manual(values=color$time_point, name="Time point") + 
    theme(axis.text.x=element_blank())

plot_3 <- ggplot(shannon[shannon$hto_demux_class=="Skin" & shannon$family_domain_subset=="Non immune cell", ], aes(x=time_point, y=shannon)) + 
    geom_boxplot(aes(fill=time_point), outlier.shape=1, outlier.size=3, size=0.5, color="black") + 
    geom_point(size=3, shape=1) +  
    ggtitle("Skin (Non immune)") + xlab("") + ylab("Shannon") +
    ylim(0, max(shannon$shannon)+0.1) +
    scale_fill_manual(values=color$time_point, name="Time point") + 
    theme(axis.text.x=element_blank())

plot_1 + plot_2 + plot_3 + patchwork::plot_layout(ncol=3, guides="collect")
```

## Time points by sample and genotype 

```r
cell_count <- dplyr::mutate(so@meta.data, group=paste(sample_name, genotype_class, family_domain_subset, hto_demux_class)) %>%
    dplyr::group_by(group, sample_name, genotype_class, family_domain_subset, hto_demux_class, time_point, cell_type_leiden_subset) %>%  
    dplyr::summarise(cell_count=n())
```

```r
cell_count_cnt <- dcast(cell_count, cell_type_leiden_subset~group, value.var="cell_count") %>% tibble::column_to_rownames("cell_type_leiden_subset") %>% replace(is.na(.), 0)
```

```r
shannon <- data.frame(group=colnames(cell_count_cnt), shannon=vegan::diversity(t(cell_count_cnt), index="simpson", base=2)) %>% 
    dplyr::left_join(., dplyr::select(cell_count, group, sample_name, genotype_class, family_domain_subset, hto_demux_class, time_point) %>% unique(), by="group") %>% 
    dplyr::group_by(genotype_class, family_domain_subset, hto_demux_class, time_point) %>% 
    dplyr::summarise(shannon_median=median(shannon), shannon_mad=mad(shannon, constant=1)) %>%
    dplyr::ungroup() %>% 
    tidyr::complete(genotype_class, family_domain_subset, hto_demux_class, time_point, fill=list(shannon_median=0, shannon_mad=0)) %>% 
    dplyr::filter(genotype_class %in% c("Host", "Donor"))
```

```r
options(repr.plot.width=10, repr.plot.height=3)

dodge_with <- 0.5
mad_scale <- 2

plot_4 <- ggplot(shannon[shannon$hto_demux_class=="Blood" & shannon$family_domain_subset=="Immune cell", ], aes(x=time_point, y=shannon_median, color=genotype_class)) + 
    geom_line(aes(group=genotype_class), size=1, position=position_dodge(dodge_with)) + 
    geom_point(shape=16, size=3, position=position_dodge(dodge_with)) + 
    geom_errorbar(aes(ymin=ifelse(shannon_median-mad_scale*shannon_mad<0, 0, shannon_median-mad_scale*shannon_mad), ymax=shannon_median+mad_scale*shannon_mad), width=0.5, size=0.75, position=position_dodge(dodge_with)) + 
    ggtitle("Blood") + xlab("") + ylab("Shannon") +
    ylim(0, max(shannon$shannon_median)+0.1) +
    scale_color_manual(values=color$genotype_class, name="Genotype class") + 
    theme(axis.text.x=element_text(angle=90, vjust=0.5, hjust=1))

plot_5 <- ggplot(shannon[shannon$hto_demux_class=="Skin" & shannon$family_domain_subset=="Immune cell", ], aes(x=time_point, y=shannon_median, color=genotype_class)) + 
    geom_line(aes(group=genotype_class), size=1, position=position_dodge(dodge_with)) + 
    geom_point(shape=16, size=3, position=position_dodge(dodge_with)) + 
    geom_errorbar(aes(ymin=ifelse(shannon_median-mad_scale*shannon_mad<0, 0, shannon_median-mad_scale*shannon_mad), ymax=shannon_median+mad_scale*shannon_mad), width=0.5, size=0.75, position=position_dodge(dodge_with)) + 
    ggtitle("Skin (Immune)") + xlab("") + ylab("Shannon") +
    ylim(0, max(shannon$shannon_median)+0.1) +
    scale_color_manual(values=color$genotype_class, name="Genotype class") + 
    theme(axis.text.x=element_text(angle=90, vjust=0.5, hjust=1))

plot_6 <- ggplot(shannon[shannon$hto_demux_class=="Skin" & shannon$family_domain_subset=="Non immune cell", ], aes(x=time_point, y=shannon_median, color=genotype_class)) + 
    geom_line(aes(group=genotype_class), size=1, position=position_dodge(dodge_with)) + 
    geom_point(shape=16, size=3, position=position_dodge(dodge_with)) + 
    geom_errorbar(aes(ymin=ifelse(shannon_median-mad_scale*shannon_mad<0, 0, shannon_median-mad_scale*shannon_mad), ymax=shannon_median+mad_scale*shannon_mad), width=0.5, size=0.75, position=position_dodge(dodge_with)) + 
    ggtitle("Skin (Non immune)") + xlab("") + ylab("Shannon") +
    ylim(0, max(shannon$shannon_median)+0.1) +
    scale_color_manual(values=color$genotype_class, name="Genotype class") + 
    theme(axis.text.x=element_text(angle=90, vjust=0.5, hjust=1))

plot_4 + plot_5 + plot_6 + patchwork::plot_layout(ncol=3, guides="collect")
```
