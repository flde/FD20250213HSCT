# Color list 

Hey Laura, short plotting example. I always create a list called color where I store the colors for my project. That works because the output of Seurat DimPlot function is just a ggplot object which you can manipulate to your likings. Check it out, it might help. If you have questions let me know

```r
# HTO demultiplex class
hto_demux_class <- c("#fcba03", "#038cfc")
names(hto_demux_class) <- c("Skin", "Blood")

# Digestion protocol 
digestion_protocol <- c("#32a852", "#a832a2")
names(digestion_protocol) <- c("old", "new")


color <- list(

    hto_demux_class=hto_demux_class, 
    digestion_protocol=digestion_protocol
    
)
```

```r
options(repr.plot.width=10, repr.plot.height=5)

dim_1 <- DimPlot(so, reduction="umap", group.by="hto_demux_class") + 
    ggtitle("HTO demux") + xlab("UMAP 1") + ylab("UMAP 2") +
    scale_color_manual(values=color$hto_demux_class) + 
    theme(aspect.ratio=1)

dim_2 <- DimPlot(so, reduction="umap", group.by="digestion_protocol") + 
    ggtitle("Digenstion protocol") + xlab("UMAP 1") + ylab("UMAP 2") +
    scale_color_manual(values=color$digestion_protocol) + 
    theme(aspect.ratio=1)

dim_1 + dim_2 + plot_layout(nrow=1, ncol=2)
```
