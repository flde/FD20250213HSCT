```r
pp_rds_file <- "data/object/qc.rds"
```

# Import data and save 

```r
so <- readRDS(pp_rds_file)
```

```r
colnames(so@meta.data)
```

```r
meta <- read.csv("data/object/components/meta.csv", row.names=1)
umap <- read.csv("data/object/components/umap.csv", row.names=1)
colnames(umap) <- c("UMAP_1", "UMAP_2")
```

```r
so@meta.data <- meta
so[["umap"]] <- CreateDimReducObject(embeddings=as.matrix(umap), key="UMAP_")
```

# Remove unwanted information for transfer 

```r
cnt <- GetAssayData(so, assay="RNA", slot="counts")
cnt <- cnt[rowSums(cnt) > 0, ]
```

```r
so_dens <- CreateSeuratObject(counts=cnt, meta.data=so@meta.data)
```

```r
so_dens[["umap"]] <- CreateDimReducObject(embeddings=as.matrix(umap), key="UMAP_")
```

# DEV 

```r
table(so_dens)
```

```r
options(repr.plot.width=2, repr.plot.height=2)

p_1 <- ggplot(so_dens@meta.data, aes(x=time_point, fill=genotype_class)) + 
    geom_bar() + 
    xlab("Time point") + ylab("Genotype [count]") + ggtitle("Genotype by SNP") + 
    scale_fill_manual(values=c("#132B43", "#56B1F7", "#8c8c8c", "#D3D3D3")) + 
    theme(legend.position="none")

p_1
```

```r
ggsave("result/figures/p_1.png", p_1, width=5, height=5, units="cm", dpi=1200)
```

```r
so_xist <- subset(so_dens, subset=time_point %in% c("baseline"))
```

```r
so_xist$patient_sex <- ifelse(so_xist$donor_sex=="female", "male", "female")
```

```r
options(repr.plot.width=2, repr.plot.height=2)

p_2 <- ggplot(so_xist@meta.data, aes(x=patient_sex, y=nXIST_RNA, fill=patient_sex)) + 
    geom_boxplot() +
    xlab("Patient sex") + ylab("XIST [counts]") + ggtitle("Genotype by XIST") +
    scale_fill_manual(values=c("#56B1F7", "#132B43")) + 
    theme(legend.position="none")

p_2
```

```r
ggsave("result/figures/p_2.png", p_2, width=5, height=5, units="cm", dpi=1200)
```
