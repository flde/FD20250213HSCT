# Cell type districution 

```r
data_1 <- dplyr::filter(so@meta.data, hto_demux_class=="Blood" & sample_name %in% n_cells_filter$sample_name) %>% 
    dplyr::group_by(sample_name, leiden_cell_type_main) %>% 
    dplyr::mutate(n_cell_type=n()) %>% 
    dplyr::group_by(sample_name) %>% 
    dplyr::mutate(library_size=n()) %>%
    dplyr::select(gvhd_class, sample_name, time_point, leiden_cell_type_main, n_cell_type, library_size) %>% 
    dplyr::ungroup() %>% 
    dplyr::mutate(cpt=n_cell_type/(library_size/10^3)) %>% 
    unique() %>% 
    dplyr::group_by(gvhd_class, time_point, leiden_cell_type_main) %>% 
    dplyr::summarise(cpt_mean=mean(cpt)) %>% 
    dplyr::group_by(gvhd_class, time_point) %>% 
    dplyr::mutate(cpt_rel=cpt_mean/sum(cpt_mean))

head(data_1)
nrow(data_1)
```

```r
options(repr.plot.width=20, repr.plot.height=5)

# bplot_1 <- ggplot(subset(so, subset=hto_demux_class=="Blood")@meta.data, aes(x=gvhd_class, fill=leiden_cell_type_main)) + 
#     geom_bar(aes(x=factor(gvhd_class, level=c("GVHD_free", "GVHD_dev"))), position="fill", width=0.75, border=NA) + 
#     ggtitle("Lymphocyte (Blood)") + xlab("") + ylab("Cell type [rel.]") + 
#     scale_fill_manual(values=color$leiden_cell_type_main_r, labels=scales::parse_format(), name="Cell type") + 
#     guides(fill=guide_legend(override.aes=list(alpha=1, size=2), ncol=1, keywidth=1, keyheight=1)) + 
#     scale_y_continuous(expand=c(0, 0.01)) + 
#     facet_grid(~time_point) + 
#     theme(
#         legend.text.align=0, 
#         axis.text.x = element_text(angle=90, vjust=0.5, hjust=1)
#     )

bplot_2 <- ggplot(data_1, aes(x=gvhd_class, y=cpt_mean, fill=leiden_cell_type_main)) + 
    geom_bar(aes(x=factor(gvhd_class, level=c("GVHD_free", "GVHD_dev"))), stat="identity", width=0.75, border=NA) + 
    ggtitle("Lymphocyte (Skin)") + xlab("") + ylab("Cell type [CPT]") + 
    scale_fill_manual(values=color$leiden_cell_type_main_r, labels=scales::parse_format(), name="Cell type") +
    guides(fill=guide_legend(override.aes=list(alpha=1, size=2), ncol=1, keywidth=1, keyheight=1)) + 
    scale_y_continuous(expand=c(0, 0.01)) + 
    facet_grid(~time_point) + 
    theme(
        legend.text.align=0, 
        axis.text.x = element_text(angle=90, vjust=0.5, hjust=1)
    )

bplot_1 + bplot_2
```

# Adjust cell type labels by tissue 

```r
so$dea_group <- factor(paste0(so$leiden_cell_type_main, paste0("*' (", so$hto_demux_class, ")'")))
```

# Import msigdbr

```r
gs_mm <- msigdbr(species="Homo sapiens", category="C2", subcategory="WIKIPATHWAYS")
gs_mm <- split(x=gs_mm$gene_symbol, f=gs_mm$gs_name)
```

# DEA

## Baseline 

```r
source("bin/seurat_dea.R")
```

```r
dea_1 <- dea_seurat(
    
    so=subset(so, subset=time_point=="Baseline"), 
    mode="compare", 
    file="result/dea/subsets/lymphocyte_gvhd_free_vs_gvhd_dev_baseline", 
    ident="dea_group",  
    map=NULL, 
    grouping_var="gvhd_class", 
    ident_1="GVHD_dev", 
    ident_2="GVHD_free", 
    only_pos=FALSE, 
    logfc_threshold=0, 
    min_pct=0, 
    compute=TRUE, 
    test_use="wilcox", 
    cnt_min=3, 
    cell_min=3, 
    gene_filter=NULL
    
)
```

```r
dea_2 <- dea_seurat(
    
    so=subset(so, subset=hto_demux_class=="Skin"), 
    mode="marker", 
    file="result/dea/subsets/lymphocyte_skin_leiden_cell_type_main", 
    ident="leiden_cell_type_main",  
    map=NULL, 
    grouping_var=NULL, 
    ident_1=NULL, 
    ident_2=NULL, 
    only_pos=FALSE, 
    logfc_threshold=0, 
    min_pct=0, 
    compute=TRUE, 
    test_use="wilcox", 
    cnt_min=3, 
    cell_min=3, 
    gene_filter=NULL
    
)
```

# GSEA

## Blood

```r
dea_1 <- dea_1[unlist(lapply(dea_1, function(x) sum(x$p_val_adj<=0.05)>=5))]
```

```r
options(repr.plot.width=2*20, repr.plot.height=ceiling(length(dea_1)/2)*10)

gsea_1 <- lapply(1:length(dea_1), function(i) gsea_from_dea(dea=dea_1[[i]], gene_sets=gs_mm, ident=names(dea_1)[i]))
gsea_plot_1 <- lapply(gsea_1, function(x) {gsea_plot(x, pval_thr=0.1, pathway_suffix="WP", color_point=setNames(c(RColorBrewer::brewer.pal(8, "Set1")[1], RColorBrewer::brewer.pal(8, "Set1")[2]), c("Adarb1KO", "Ctr")), size_range_max=5) + theme_global_set(size_select=1)})
wrap_plots(gsea_plot_1, ncol=2)
```
