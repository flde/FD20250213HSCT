```r
source <- c("Fibroblasts")
target <- c("T['h']*' RM'")
sample_group <- c("Baseline"="Baseline", "Tx"="Tx", "HSCT"="HSCT", "GVHD"="GVHD")
```

```r
lr_res_i <- lr_res_parse(lr_res, source, target, sample_group, pval_prob_thr) %>% dplyr::mutate(col_name=factor(col_name, levels=c("Baseline", "Tx", "HSCT", "GVHD")))
lr_res_i <- lr_res_stretch(lr_res_i)
```
