# Convert cellSNP output to vartrix format and filter minMAF

```r
cellsnp_out_dir <- gsub("_cellsnp_0", "_vartrix_0", cellsnp_in_dir)
catch <- lapply(1:length(cellsnp), function(i) {vartrix_cellsnp_vartrix(cellsnp[[i]], cellsnp_out_dir[[i]], minMAF=NULL)})

cellsnp_out_dir <- gsub("_cellsnp_0", "_vartrix_0001", cellsnp_in_dir)
catch <- lapply(1:length(cellsnp), function(i) {vartrix_cellsnp_vartrix(cellsnp[[i]], cellsnp_out_dir[[i]], minMAF=0.0001)})

cellsnp_out_dir <- gsub("_cellsnp_0", "_vartrix_0005", cellsnp_in_dir)
catch <- lapply(1:length(cellsnp), function(i) {vartrix_cellsnp_vartrix(cellsnp[[i]], cellsnp_out_dir[[i]], minMAF=0.0005)})

cellsnp_out_dir <- gsub("_cellsnp_0", "_vartrix_001", cellsnp_in_dir)
catch <- lapply(1:length(cellsnp), function(i) {vartrix_cellsnp_vartrix(cellsnp[[i]], cellsnp_out_dir[[i]], minMAF=0.001)})

cellsnp_out_dir <- gsub("_cellsnp_0", "_vartrix_005", cellsnp_in_dir)
catch <- lapply(1:length(cellsnp), function(i) {vartrix_cellsnp_vartrix(cellsnp[[i]], cellsnp_out_dir[[i]], minMAF=0.005)})

cellsnp_out_dir <- gsub("_cellsnp_0", "_vartrix_01", cellsnp_in_dir)
catch <- lapply(1:length(cellsnp), function(i) {vartrix_cellsnp_vartrix(cellsnp[[i]], cellsnp_out_dir[[i]], minMAF=0.01)})

cellsnp_out_dir <- gsub("_cellsnp_0", "_vartrix_05", cellsnp_in_dir)
catch <- lapply(1:length(cellsnp), function(i) {vartrix_cellsnp_vartrix(cellsnp[[i]], cellsnp_out_dir[[i]], minMAF=0.05)})

cellsnp_out_dir <- gsub("_cellsnp_0", "_vartrix_1", cellsnp_in_dir)
catch <- lapply(1:length(cellsnp), function(i) {vartrix_cellsnp_vartrix(cellsnp[[i]], cellsnp_out_dir[[i]], minMAF=0.1)})

cellsnp_out_dir <- gsub("_cellsnp_0", "_vartrix_2", cellsnp_in_dir)
catch <- lapply(1:length(cellsnp), function(i) {vartrix_cellsnp_vartrix(cellsnp[[i]], cellsnp_out_dir[[i]], minMAF=0.2)})
```
