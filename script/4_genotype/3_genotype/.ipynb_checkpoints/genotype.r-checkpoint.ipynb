# Import 

```r
so <- readRDS("data/object/pp.rds")
genotype <- read.csv("result/genotype/vireo_mode_1a_genome1k.csv", row.names=1)
```

# Filter genotype by cell_id 

```r
genotype <- genotype[genotype$cell_id %in% so$cell_id, ]
```

```r
# Set factor levels 
genotype$genotype_class <- factor(genotype$genotype_class, levels=c("Host", "Donor", "unassigned", "doublet"))
genotype$genotype_best_class <- factor(genotype$genotype_best_class, levels=c("Host", "Donor"))
genotype$patient_id <- factor(genotype$patient_id, levels=paste0("Patient_", 1:11))
```

# Time point sample comparison 

```r
data <- genotype %>% 
    dplyr::filter(patient_id %in% paste0("Patient_", c(1, 3:7))) %>%
    dplyr::filter(time_point %in% c("Baseline", "Tx")) %>% 
    dplyr::group_by(patient_id, minmaf, genotype_class) %>% summarise(n=n()) %>% 
    dplyr::group_by(patient_id, minmaf) %>% mutate(ratio=n/sum(n))

plot_1 <-ggplot(data, aes(x=as.character(minmaf), y=ratio, fill=genotype_class)) + 
    geom_bar(stat="identity") + 
    ggtitle("Basline/Tx samples") + xlab("minMAF") + ylab("ratio") + 
    scale_fill_manual("Vireo", values=setNames(c("#132B43", "#56B1F7", "#8c8c8c", "#ffc425"), c("Host", "Donor", "doublet", "unassigned"))) + 
    facet_grid(~patient_id) + 
    theme(axis.text.x=element_text(angle=45, vjust=1, hjust=1))
```

```r
data <- genotype %>% 
    dplyr::filter(patient_id %in% paste0("Patient_", c(1, 3:7))) %>%
    dplyr::filter(time_point %in% c("Baseline", "Tx")) %>%
    droplevels() %>% 
    tidyr::complete(patient_id, genotype_class, minmaf, fill=list(pXIST_RNA=0), explicit=TRUE)

plot_2 <-ggplot(data, aes(x=minmaf, y=pXIST_RNA, fill=genotype_class, color=genotype_class)) + 
    geom_boxplot(position=position_dodge2(preserve="single"), outlier.size=0.5) + 
    ggtitle("Basline/Tx samples") + 
    scale_color_manual("Vireo", values=setNames(c("#132B43", "#56B1F7", "#8c8c8c", "#ffc425"), c("Host", "Donor", "unassigned", "doublet")), drop=FALSE) + 
    scale_fill_manual("Vireo", values=setNames(c("#132B43", "#56B1F7", "#8c8c8c", "#ffc425"), c("Host", "Donor", "unassigned", "doublet")), drop=FALSE) + 
    facet_grid(~patient_id,) + 
    theme(axis.text.x=element_text(angle=45, vjust=1, hjust=1))
```

# Get best result (vartrix 0.01) 

```r
genotype_vartrix_01 <- genotype[genotype$minmaf=="vartrix 0.01", ]
```

```r
# Set genotype of Patient 2 to Host 
genotype_vartrix_01 <- genotype_vartrix_01 %>% dplyr::mutate(genotype_class=if_else(patient_id=="Patient_2", "Host", genotype_class))
genotype_vartrix_01 <- genotype_vartrix_01 %>% dplyr::mutate(genotype_best_class=if_else(patient_id=="Patient_2", "Host", genotype_best_class))
```

```r
# Set genotype of Patient 9 to NA
genotype_vartrix_01 <- genotype_vartrix_01 %>% dplyr::mutate(genotype_class=if_else(patient_id=="Patient_9", NA, genotype_class))
genotype_vartrix_01 <- genotype_vartrix_01 %>% dplyr::mutate(genotype_best_class=if_else(patient_id=="Patient_9", NA, genotype_best_class))
```

```r
genotype_vartrix_01$genotype_class <- factor(genotype_vartrix_01$genotype_class, levels=c("Host", "Donor", "doublet", "unassigned"))
genotype_vartrix_01$genotype_best_class <- factor(genotype_vartrix_01$genotype_best_class, levels=c("Host", "Donor"))
genotype_vartrix_01$patient_id <- factor(genotype_vartrix_01$patient_id, levels=paste0("Patient_", 1:11))
```

```r
stat <- genotype_vartrix_01 %>% 
    dplyr::filter(patient_id %in% paste0("Patient_", c(1, 3:11))) %>%
    dplyr::filter(time_point %in% c("Baseline", "Tx"))

rbind(table(stat$genotype_class), c(table(stat$genotype_best_class), NA, NA))
```

```r
stat <- genotype_vartrix_01 %>% 
    dplyr::filter(patient_id %in% paste0("Patient_", c(1, 3:11))) %>%
    dplyr::filter(cell_type_domain %in% c("non-immune cell"))

rbind(table(stat$genotype_class), c(table(stat$genotype_best_class), NA, NA))
```

```r
# Set genotype of baseline/tx samples 
genotype_vartrix_01 <- genotype_vartrix_01 %>% dplyr::mutate(genotype_class=if_else(time_point %in% c("Baseline", "Tx") & genotype_class=="Donor", NA, genotype_class))
genotype_vartrix_01 <- genotype_vartrix_01 %>% dplyr::mutate(genotype_best_class=if_else(time_point %in% c("Baseline", "Tx") & genotype_best_class=="Donor", NA, genotype_best_class))
```

```r
# Set genotype of structural cells 
genotype_vartrix_01 <- genotype_vartrix_01 %>% dplyr::mutate(genotype_class=if_else(cell_type_domain %in% c("non-immune cell") & genotype_class=="Donor", NA, genotype_class))
genotype_vartrix_01 <- genotype_vartrix_01 %>% dplyr::mutate(genotype_best_class=if_else(cell_type_domain %in% c("non-immune cell") & genotype_best_class=="Donor", NA, genotype_best_class))
```

```r
stat <- genotype_vartrix_01 %>% 
    dplyr::filter(patient_id %in% paste0("Patient_", c(1, 3:11))) %>%
    dplyr::filter(time_point %in% c("Baseline", "Tx"))

rbind(table(stat$genotype_class), c(table(stat$genotype_best_class), NA, NA))
```

```r
stat <- genotype_vartrix_01 %>% 
    dplyr::filter(patient_id %in% paste0("Patient_", c(1, 3:11))) %>%
    dplyr::filter(cell_type_domain %in% c("non-immune cell"))

rbind(table(stat$genotype_class), c(table(stat$genotype_best_class), NA, NA))
```
