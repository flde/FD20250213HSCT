```r
net_analysis_signaling_role_hm <- function(data_1, data_2, group, cell_type, pathway_select=NULL, show_heatmap_legend=FALSE) {
    
    mat <- dplyr::full_join(data_1[, c("pathway", cell_type)], data_2[, c("pathway", cell_type)], by=join_by(pathway)) %>% 
        tibble::column_to_rownames("pathway") %>% 
        dplyr::filter(rowSums(is.na(.)) < ncol(.)) %>% 
        replace(is.na(.), 0)
    
    if(!is.null(pathway_select)) {mat <- mat[pathway_select, ]}

    colnames(mat) <- group

    color_ramp_mat <- c("#ffffff", rev(RColorBrewer::brewer.pal(11,"RdBu"))[11])
    breaks_mat <- seq(0, 1, length.out=length(color_ramp_mat))
    color_function_mat <- circlize::colorRamp2(breaks_mat, color_ramp_mat)

    ComplexHeatmap::Heatmap(
        
        matrix=mat,

        col=color_function_mat, 
        na_col="white", 

        width=unit(12*ncol(mat), "mm"), 
        height=unit(6*nrow(mat), "mm"), 

        # row_title=NULL, 
        row_title_gp=gpar(fontsize=16, fontface="bold"), 

        column_title=parse(text=cell_type), 
        column_title_gp=gpar(fontsize=18, fontface="bold"), 

        row_names_gp=gpar(fontsize=16, fontface="italic"), 
        column_names_gp=gpar(fontsize=16, fontface="plain"), 

        cluster_rows=TRUE, 
        cluster_row_slices=FALSE, 
        show_row_dend=FALSE,
        # row_split=row_split, 
        row_gap=unit(2, "mm"), 
        show_row_names=TRUE, 
        row_names_side="left", 

        cluster_columns=FALSE, 
        cluster_column_slices=FALSE, 
        show_column_dend=FALSE, 
        # column_split=column_split_idx, 
        column_gap=unit(1, "mm"), 
        show_column_names=TRUE, 

        # top_annotation=top_annotation, 
        left_annotation=NULL, 

        rect_gp=gpar(col="black", lwd=0, alpha=1), 

        heatmap_legend_param=list(title="Relative strength", title_gp=gpar(fontsize=16, fontface="plain"), labels_gp=gpar(fontsize=16), grid_width=unit(2, "mm"), legend_width=unit(2, "mm"), grid_height=unit(10, "mm"), legend_height=unit(10, "mm"), at=breaks_mat), 
        # show_heatmap_legend=ifelse(idx==1, TRUE, FALSE),

        border=TRUE, 
        border_gp=gpar(col="black", lwd=unit(1, "mm")), 
        
        show_heatmap_legend=show_heatmap_legend, 

        use_raster=FALSE, raster_by_magick=FALSE

    )

}
```

```r
net_probability_hm <- function(data_1, data_2, group, pathway_select, pattern, cell_type, show_heatmap_legend=FALSE) {    
    
    data_1 <- do.call("rbind", data_1)
    data_2 <- do.call("rbind", data_2)

    mat <- rbind(dplyr::mutate(data_1, group=group[1]), dplyr::mutate(data_2, group=group[2]))
    
    mat <- na.omit(mat)

    mat <- dplyr::mutate(mat, ligand_receptor=paste0(ligand, " > ", receptor))

    if(!is.null(pathway_select)) {mat <- dplyr::filter(mat, pathway_name %in% pathway_select)}

    if(pattern=="outgoing") {mat <- mat %>% dplyr::group_by(group, ligand_receptor, source) %>% dplyr::mutate(prob=mean(prob), pval=mean(pval)) %>% dplyr::select(group, pathway_name, ligand_receptor, source, prob, pval)}
    if(pattern=="incoming") {mat <- mat %>% dplyr::group_by(group, ligand_receptor, target) %>% dplyr::mutate(prob=mean(prob), pval=mean(pval)) %>% dplyr::select(group, pathway_name, ligand_receptor, target, prob, pval)}

    mat <- mat %>% dplyr::ungroup() %>% dplyr::distinct()

    if(pattern=="outgoing") {mat <- reshape2::dcast(mat, pathway_name + ligand_receptor + source ~ group, value.var="prob") %>% replace(is.na(.), 0)}
    if(pattern=="incoming") {mat <- reshape2::dcast(mat, pathway_name + ligand_receptor + target ~ group, value.var="prob") %>% replace(is.na(.), 0)}

    colnames(mat)[3] <- "cell_type"

    if("Tx" %in% group) {mat <- dplyr::group_by(mat, pathway_name, ligand_receptor) %>% tidyr::complete(., cell_type, fill=list("Baseline"=0, "Tx"=0))}
    if("GVHD" %in% group) {mat <- dplyr::group_by(mat, pathway_name, ligand_receptor) %>% tidyr::complete(., cell_type, fill=list("Baseline"=0, "GVHD"=0))}

    mat <- dplyr::filter(mat, cell_type=={{cell_type}})
    mat <- dplyr::arrange(mat, pathway_name, ligand_receptor)

    # Get data 
    row_split <- mat[, 1, drop=TRUE]
    mat <- mat[, c(2, 4, 5)] %>% tibble::column_to_rownames("ligand_receptor")

    color_ramp_mat <- c("#ffffff", rev(RColorBrewer::brewer.pal(11,"RdBu"))[11])
    breaks_mat <- seq(min(mat), max(mat), length.out=length(color_ramp_mat))
    color_function_mat <- circlize::colorRamp2(breaks_mat, color_ramp_mat)

    ComplexHeatmap::Heatmap(
        
        matrix=mat,

        col=color_function_mat, 
        na_col="white", 

        width=unit(12*ncol(mat), "mm"), 
        height=unit(6*nrow(mat), "mm"), 

        # row_title=NULL, 
        row_title_gp=gpar(fontsize=16, fontface="bold"), 

        column_title=parse(text=cell_type), 
        column_title_gp=gpar(fontsize=18, fontface="bold"), 

        row_names_gp=gpar(fontsize=16, fontface="italic"), 
        column_names_gp=gpar(fontsize=16, fontface="plain"), 

        cluster_rows=TRUE, 
        cluster_row_slices=FALSE, 
        show_row_dend=FALSE,
        row_split=row_split, 
        row_gap=unit(2, "mm"), 
        show_row_names=TRUE, 
        row_names_side="right", 
        row_title_rot=0, 

        cluster_columns=FALSE, 
        cluster_column_slices=FALSE, 
        show_column_dend=FALSE, 
        # column_split=column_split_idx, 
        column_gap=unit(1, "mm"), 
        show_column_names=TRUE, 

        # top_annotation=top_annotation, 
        left_annotation=NULL, 

        rect_gp=gpar(col="black", lwd=0, alpha=1), 

        heatmap_legend_param=list(title="Relative strength", title_gp=gpar(fontsize=16, fontface="plain"), labels_gp=gpar(fontsize=16), grid_width=unit(2, "mm"), legend_width=unit(2, "mm"), grid_height=unit(10, "mm"), legend_height=unit(10, "mm"), at=breaks_mat), 
        # show_heatmap_legend=ifelse(idx==1, TRUE, FALSE),

        border=TRUE, 
        border_gp=gpar(col="black", lwd=unit(1, "mm")), 
        
        show_heatmap_legend=show_heatmap_legend, 

        use_raster=FALSE, raster_by_magick=FALSE

    )

}
```
