# Filter by number of cells per cell type 

## Number of cells per sample and group 

```r
n_cells <- dplyr::group_by(so@meta.data, sample_name, time_point, hto_demux_class, gvhd_class) %>% 
    dplyr::summarise(n_cells=n())
```

```r
options(repr.plot.width=6, repr.plot.height=5)

ggplot(n_cells, aes(x=time_point, y=n_cells)) + 
    geom_boxplot() + geom_point(aes(color=gvhd_class), size=2.5) + 
    ggtitle("Cells per sample") + ylab("Cells [Count]") + xlab("") + 
    scale_color_manual(values=c("#ffc425", "#56B1F7"), name="GvHD class") + 
    facet_grid(~hto_demux_class) + 
    theme(axis.text.x=element_text(angle=90, vjust=0.5, hjust=1))
```

## Number of cells per cell type and group

```r
n_cells <- dplyr::group_by(so@meta.data, gvhd_class, hto_demux_class, time_point, cell_type_leiden_subset) %>% 
    dplyr::summarise(n_cells=n()) %>% 
    dplyr::filter(time_point %in% c("Baseline", "D14", "D100")) %>% dplyr::ungroup() %>% as.data.frame()
```

```r
n_cells <-dplyr::full_join(
    
    n_cells[n_cells$gvhd_class=="GVHD_dev", ] %>% dplyr::rename(n_cells_gvhd_dev="n_cells") %>% dplyr::select(-gvhd_class), 
    n_cells[n_cells$gvhd_class=="GVHD_free", ] %>% dplyr::rename(n_cells_gvhd_free="n_cells") %>% dplyr::select(-gvhd_class), 
    by=c("hto_demux_class", "time_point", "cell_type_leiden_subset")

) %>% dplyr::filter(is.na(cell_type_leiden_subset)==FALSE) %>% replace(is.na(.), 0)
```

```r
write.csv(n_cells[n_cells$hto_dmux_class=="Blood", ], "result/dea/gvhd_class_cell_type_count_blood.csv")
write.csv(n_cells[n_cells$hto_dmux_class=="Skin", ], "result/dea/gvhd_class_cell_type_count_skin.csv")
```

## Filter groups for at least 10 cells in each group 

```r
cell_min <- 10
```

```r
exclude <- n_cells[n_cells$n_cells_gvhd_dev < cell_min | n_cells$n_cells_gvhd_free < cell_min, ]
nrow(exclude)
```

```r
include <- n_cells[n_cells$n_cells_gvhd_dev >= cell_min & n_cells$n_cells_gvhd_free >= cell_min, ]
nrow(include)
```

```r
# Set sample group for inclution of comparisons 
include$sample_group <- paste0(include$hto_demux_class, "_", include$time_point, "_", include$cell_type_leiden_subset)
so$sample_group <- paste0(so$hto_demux_class, "_", so$time_point, "_", so$cell_type_leiden_subset)
```

```r
so <- subset(so, subset=sample_group %in% include$sample_group)
```

# Helper function 

```r
set_order <- function(dea) {
    
    order <- names(cell_type_leiden_subset_r)[names(cell_type_leiden_subset_r) %in% names(dea)]
    dea <- dea[order]
    
    return(dea)

}
```

# GSEA

```r
gs_mm <- msigdbr(species="Homo sapiens", category="C2", subcategory="WIKIPATHWAYS")
gs_mm <- split(x=gs_mm$gene_symbol, f=gs_mm$gs_name)
```

## Baseline (Blood)

```r
options(repr.plot.width=2*20, repr.plot.height=ceiling(length(dea_1)/2)*10)

gsea_1 <- lapply(1:length(dea_1), function(i) gsea_from_dea(dea=dea_1[[i]], gene_sets=gs_mm, ident=names(dea_1)[i]))
gsea_plot_1 <- lapply(gsea_1, function(x) {gsea_plot(x, pval_thr=0.1, pathway_suffix="WP", color_point=setNames(c(RColorBrewer::brewer.pal(8, "Set1")[1], RColorBrewer::brewer.pal(8, "Set1")[2]), c("Adarb1KO", "Ctr")), size_range_max=5) + theme_global_set(size_select=1)})
wrap_plots(gsea_plot_1, ncol=2)
```
