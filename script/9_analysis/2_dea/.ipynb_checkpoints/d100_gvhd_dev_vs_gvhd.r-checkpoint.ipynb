# DEA

## D100 GVHD dev vs GVHD (Blood)

```r
dea_1 <- dea_seurat(
    
    so=subset(so, subset=((time_point=="D100" & gvhd_class=="GVHD_dev") | (time_point=="GVHD")) & hto_demux_class=="Blood"), 
    mode="limma_voom_re", 
    file="result/dea/comparison/d100_gvhd_dev_vs_gvhd_blood", 
    ident="cell_type_leiden_subset",  
    map=NULL, 
    grouping_var="time_point", 
    ident_1="GVHD", 
    ident_2="D100", 
    random_effect_var="patient_id", 
    only_pos=FALSE, 
    logfc_threshold=0, 
    min_pct=0, 
    compute=TRUE, 
    test_use="wilcox", 
    cnt_min=3, 
    cell_min=3
    
)
```

## D100 GVHD dev vs GVHD (Skin)

```r
dea_2 <- dea_seurat(
    
    so=subset(so, subset=((time_point=="D100" & gvhd_class=="GVHD_dev") | (time_point=="GVHD")) & hto_demux_class=="Skin"), 
    mode="limma_voom_re", 
    file="result/dea/comparison/d100_gvhd_dev_vs_gvhd_skin", 
    ident="cell_type_leiden_subset",  
    map=NULL, 
    grouping_var="time_point", 
    ident_1="GVHD", 
    ident_2="D100", 
    random_effect_var="patient_id", 
    only_pos=FALSE, 
    logfc_threshold=0, 
    min_pct=0, 
    compute=TRUE, 
    test_use="wilcox", 
    cnt_min=3, 
    cell_min=3
    
)
```

# Volcano plots

## D100 GVHD dev vs GVHD (Blood)

```r
options(repr.plot.width=4*8, repr.plot.height=ceiling(length(dea_1)/4)*8)

vp_treatment <- lapply(seq_along(dea_1), function(i) vp_dea(dea_1[[i]], title=base::str2expression(as.character(dea_1[[i]]$ident[1])), log2_thold=0, adjpvalue_thold=0.1) + theme_global_set(size_select=1))
wrap_plots(vp_treatment, ncol=4, nrow=ceiling(length(dea_1)/4))
```

## D100 GVHD dev vs GVHD (Skin)

```r
options(repr.plot.width=4*8, repr.plot.height=ceiling(length(dea_2)/4)*8)

vp_treatment <- lapply(seq_along(dea_2), function(i) vp_dea(dea_2[[i]], title=base::str2expression(as.character(dea_2[[i]]$ident[1])), log2_thold=0, adjpvalue_thold=0.1) + theme_global_set(size_select=1))
wrap_plots(vp_treatment, ncol=4, nrow=ceiling(length(dea_2)/4))
```

# GSEA

```r
gs_mm <- msigdbr(species="Homo sapiens", category="C2", subcategory="WIKIPATHWAYS")
gs_mm <- split(x=gs_mm$gene_symbol, f=gs_mm$gs_name)
```

## D100 GVHD dev vs GVHD (Blood)

```r
options(repr.plot.width=2*20, repr.plot.height=ceiling(length(dea_1)/2)*10)

gsea_1 <- lapply(1:length(dea_1), function(i) gsea_from_dea(dea=dea_1[[i]], gene_sets=gs_mm, ident=names(dea_1)[i]))
gsea_plot_1 <- lapply(gsea_1, function(x) {gsea_plot(x, pval_thr=0.1, pathway_suffix="WP", color_point=setNames(c(RColorBrewer::brewer.pal(8, "Set1")[1], RColorBrewer::brewer.pal(8, "Set1")[2]), c("GVHD", "D100 GVHD dev")), size_range_max=5) + theme_global_set(size_select=1)})
wrap_plots(gsea_plot_1, ncol=2)
```

## D100 GVHD dev vs GVHD (Skin)

```r
options(repr.plot.width=2*20, repr.plot.height=ceiling(length(dea_2)/2)*10)

gsea_2 <- lapply(1:length(dea_2), function(i) gsea_from_dea(dea=dea_2[[i]], gene_sets=gs_mm, ident=names(dea_2)[i]))
gsea_plot_2 <- lapply(gsea_2, function(x) {gsea_plot(x, pval_thr=0.1, pathway_suffix="WP", color_point=setNames(c(RColorBrewer::brewer.pal(8, "Set1")[1], RColorBrewer::brewer.pal(8, "Set1")[2]), c("GVHD", "D100 GVHD dev")), size_range_max=5) + theme_global_set(size_select=1)})
wrap_plots(gsea_plot_2, ncol=2)
```
