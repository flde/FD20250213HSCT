# Residency vs circulating score 

## CD4+ 

```python
adata_0 = sc.read_h5ad('data/object/pseudotime/scvi_pt_0.h5ad')
```

```python
diffTscore = pd.read_csv('script/figure/figure_4/data/diffTscore.csv', index_col=0)
```

```python
adata_0.obs = adata_0.obs.merge(diffTscore, left_index=True, right_index=True, how='left')
```

```python
adata_0.obs['dummy'] = 'dummy'
adata_0.uns['dummy_colors'] = ['#D4D4D4']
```

```python
fig, ax = plt.subplots()

sc.pl.umap(adata_0, color=['dummy'], add_outline=True, legend_loc="none", show=False, ax=ax)
scv.pl.paga(adata_0, basis="umap", threshold=0.25, fontsize=8, node_size_scale=0.75, node_size_power=1.0, max_edge_width=0.5, color=['diffTscore'], legend_loc='on data', show=False, ax=ax)

plt.savefig('result/figures/figure_4/paga_cd4pos_residency_vs_recirculating_scores.pdf', dpi=300)
```
