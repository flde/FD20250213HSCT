# Parameter settings

```r
# Seurat object 
pp_rds_file <- "data/object/pp.rds"

# Meta data 
cell_type_leiden_file <- "result/annotation/cell_type_leiden.txt"
genotype_file <- "result/genotype/genotype_mode_1a_genome1k.csv"
scirpy_file <- "result/scirpy/scirpy.csv"
```

# Import Seurat object

```r
so <- readRDS(pp_rds_file)
```

## Update meta data 

```r
so <- AddMetaData(so, read.delim(cell_type_leiden_file, row.names=1)[, 1, drop=FALSE])
so <- AddMetaData(so, read.csv(scirpy_file, row.names=1)[c("has_ir", "cc_aa_alignment", "cc_aa_alignment_size")])
so <- AddMetaData(so, read.csv(genotype_file, row.names="cell_id")[c("genotype_class")])
```

## Set factor levels 

```r
set_factor_levels <- function(so) {
    
    so$hto_demux_class <- factor(so$hto_demux_class, levels=names(color$hto_demux_class))
    so$genotype_class <- factor(so$genotype_class, levels=names(color$genotype_class))
    
    return(so)
}
```

```r
so <- set_factor_levels(so)
```

# UMAP

```r
options(repr.plot.width=4*10, repr.plot.height=10)
source("bin/seurat_qc.R")

dplot_1 <- dplot(so, reduction="umap", group_by="cell_type_leiden", alpha=1, pt_size=1.5) + ggtitle("Leiden annotation")

dplot_2 <- dplot(so, reduction="umap", group_by="hto_demux_class", alpha=1, pt_size=1.5) + ggtitle("Tissue") + scale_color_manual(values=color$hto_demux_class, na.value="grey50")

dplot_3 <- dplot(subset(so, subset=genotype_class %in% c("Host", "Donor", "unassigned", "doublet")), reduction="umap", group_by="genotype_class", alpha=1, pt_size=1.5) + ggtitle("Genotype") + scale_color_manual(values=color$genotype_class, na.value="grey50")

fplot_4 <- fplot(so, reduction="umap", features="cc_aa_alignment_size", alpha=1, pt_size=1.5) + ggtitle("TCR expansion") 

dplot_1 + dplot_2 + dplot_3 + fplot_4 + plot_layout(ncol=4)
```
