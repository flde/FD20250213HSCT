# CellAssign 

## CD4+ 

```python
adata = sc.read_h5ad('data/object/pseudotime/scvi_pt_0.h5ad')
```

```python
lib_size = adata.X.sum(1)
adata.obs["size_factor"] = lib_size / np.mean(lib_size)
```

```python
# Define gene names and cell types
genes = ['CCR6', 'IL7R', 'IFNG', 'CXCR3']
cell_types = ['pos', 'pos', 'neg', 'neg']

# Create a dictionary to initialize the DataFrame
data = {'pos': [1 if cell_type == 'pos' else 0 for cell_type in cell_types],
        'neg': [1 if cell_type == 'neg' else 0 for cell_type in cell_types]}

# Create the DataFrame
marker_gene_mat = pd.DataFrame(data, index=genes)
```

```python
adata_ca  = adata[:, marker_gene_mat.index].copy()
```

```python
scvi.external.CellAssign.setup_anndata(adata_ca, size_factor_key="size_factor")
```

```python
model = CellAssign(adata_ca, marker_gene_mat)
model.train()
```

```python
model.history["elbo_validation"].plot()
```

```python
predictions = model.predict()
```

```python
adata.obs["predictions"] = predictions.idxmax(axis=1).values
```

```python
adata.obs.loc[list(predictions.max(axis=1) < 0.99), 'predictions']=None
```

```python
sc.pp.normalize_total(adata)
sc.pp.log1p(adata)
```

```python
sc.pl.umap(adata, color=["predictions", 'CCR6', 'IL7R', 'MAF'], frameon=False, ncols=4, size=50, use_raw=False)
```
