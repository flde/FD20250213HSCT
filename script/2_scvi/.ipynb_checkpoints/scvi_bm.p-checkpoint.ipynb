# Loading and preparing data 

```python
import scvi
import scanpy as sc

import os
```

```python
sc.set_figure_params(figsize=(5, 5))
```

```python
os.chdir('/research/peer/fdeckert/FD20210520LEUKO')
```

# Import AnnData

```python
adata = sc.read_h5ad('data/object/qc.h5ad')
```

```python
adata=adata[adata.obs['tissue']=="BM"].copy()
adata=adata[adata.obs['solo_label']=='singlet'].copy()
sc.pp.filter_genes(adata, min_counts=1)
```

# Register model

```python
adata=scvi.data.setup_anndata(
    adata,
    categorical_covariate_keys=['treatment'],
    copy=True
)
```

# Creating and training a model 

```python
model=scvi.model.SCVI(
    adata, 
    n_latent=10, 
    n_hidden=128, 
    n_layers=2, 
    gene_likelihood='nb', 
    dispersion='gene'
)
```

```python
model.train()
```

```python
model.save('data/object/bm/model_drmv_l10/', overwrite=True)
```

# Update anndata object with scvi results 

```python
# scvi normalized counts 
adata.layers['data']=model.get_normalized_expression(library_size=10e4)
adata.obsm['latent']=model.get_latent_representation()
```

# Leiden cluster and UMAP

```python
sc.pp.neighbors(adata, n_neighbors=10, n_pcs=30, use_rep='latent')
sc.tl.leiden(adata)
sc.tl.umap(adata)
```

```python
sc.pl.umap(adata, color=['leiden', 'treatment', 'tissue', 'cc_phase_class', 'label_main_immgen', 'solo_pred_doublet', 'solo_pred_singlet', 'solo_label'], frameon=False, ncols=3, wspace=0.5)
```

# Store results 

```python
adata.write('data/object/bm/andata_scvi_drmv_l10.h5ad')
```
