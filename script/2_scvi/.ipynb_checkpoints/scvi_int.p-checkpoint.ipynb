```python
import scvi
import scanpy as sc

import pandas as pd
import numpy as np
import scipy

import os
import sys

sc.set_figure_params(figsize=(5, 5))
```

# Import custom modules

```python
sys.path.append('/research/peer/fdeckert/scFacility/script/')
import dirFacility
```

# Import AnnData object

```python
adata = dirFacility.dir2adata('/research/peer/fdeckert/FD20210520LEUKO/data/object/so_qc/', assay='RNA', slot='counts')
```

# Compute highly variable genes

```python
hvf_1_1 = pd.read_csv('data/object/bm_pbs/assay/SCT/variable_features.csv').iloc[:, 0].tolist()
hvf_1_2 = pd.read_csv('data/object/bm_pic/assay/SCT/variable_features.csv').iloc[:, 0].tolist()
hvf_1_3 = pd.read_csv('data/object/bm_pbs_lps/assay/SCT/variable_features.csv').iloc[:, 0].tolist()
hvf_1_4 = pd.read_csv('data/object/bm_pic_lps/assay/SCT/variable_features.csv').iloc[:, 0].tolist()

hvf_1 = list(set(hvf_1_1) | set(hvf_1_2) | set(hvf_1_3) | set(hvf_1_4))
```

```python
hvf_2_1 = pd.read_csv('data/object/pbmc_pbs/assay/SCT/variable_features.csv').iloc[:, 0].tolist()
hvf_2_2 = pd.read_csv('data/object/pbmc_pic/assay/SCT/variable_features.csv').iloc[:, 0].tolist()
hvf_2_3 = pd.read_csv('data/object/pbmc_pbs_lps/assay/SCT/variable_features.csv').iloc[:, 0].tolist()
hvf_2_4 = pd.read_csv('data/object/pbmc_pic_lps/assay/SCT/variable_features.csv').iloc[:, 0].tolist()

hvf_2 = list(set(hvf_2_1) | set(hvf_2_2) | set(hvf_2_3) | set(hvf_2_4))
```

```python
hvf = list(set(hvf_1) | set(hvf_2))
```

```python
adata = adata[:, adata.var_names.isin(hvf)].copy()
```

```python
adata.layers['counts'] = adata.X.copy() # preserve counts
```

# Register model

```python
adata=scvi.data.setup_anndata(
    adata,
    layer='counts',
    categorical_covariate_keys=['treatment'], 
    continuous_covariate_keys=['msCC_diff_RNA'], 
    copy=True
)
```

# Creating and training a model 

```python
model=scvi.model.SCVI(
    adata, 
    n_latent=10, 
    n_hidden=128, 
    n_layers=2, 
    dispersion='gene'
)
```

```python
model.train()
```

```python
model.save('data/object/so_qc/model/', overwrite=True)
```

# Update anndata object with scvi results 

```python
# scvi normalized counts 
adata.layers['data']=model.get_normalized_expression(library_size=10e4)
```

```python
adata.obsm['latent']=model.get_latent_representation()
```

# Store results 

```python
adata.write('data/object/so_qc/andata.h5ad')
```

```python
dirFacility.adata2dir(adata, 'data/object/so_qc/', assay='SCVI', slot='data', build_dir=False, overwrite=True)
```
