# Import scvi as Seurat Object

```r
so_bm <- dir2seurat(dir="data/object/bm", assays="SCVI", slot="counts")
so_bm[["RNA"]] <- so_bm[["SCVI"]]
so_blood <- dir2seurat(dir="data/object/blood", assays="SCVI", slot="counts")
so_blood[["RNA"]] <- so_blood[["SCVI"]]
so_comb <- dir2seurat(dir="data/object/comb", assays="SCVI", slot="counts")
so_comb[["RNA"]] <- so_comb[["SCVI"]]
```

```r
DefaultAssay(so_bm) <- "SCVI"
DefaultAssay(so_blood) <- "SCVI"
DefaultAssay(so_comb) <- "SCVI"
```

# Latent space dimensional reduction and clustering

```r
# # Find clusters, then run UMAP, and visualize
so_comb <- FindNeighbors(so_comb, dims=1:10, reduction="latent", assay="SCVI", graph.name=c("SCVI_nn", "SCVI_snn"), return.neighbor=FALSE, verbose=FALSE)
so_comb <- FindClusters(so_comb, algorithm=4, graph.name="SCVI_snn", random.seed=random_seed, method="igraph", verbose=FALSE) # Leiden will not work on 40k + cells only if method is igraph
so_comb <- RunUMAP(so_comb, dims=1:10, reduction="latent", umap.method="umap-learn", seed.use=random_seed, verbose=FALSE)
```

```r
saveRDS(so_comb, "data/object/comb_scvi.rds")
```

```r
# Find clusters, then run UMAP, and visualize
so_bm <- FindNeighbors(so_bm, dims=1:10, reduction="latent", assay="SCVI", graph.name=c("SCVI_nn", "SCVI_snn"), return.neighbor=FALSE, verbose=FALSE)
so_bm <- FindClusters(so_bm, algorithm=4, graph.name="SCVI_snn", random.seed=random_seed, verbose=FALSE)
so_bm <- RunUMAP(so_bm, dims=1:10, reduction="latent", umap.method="umap-learn", seed.use=random_seed, verbose=FALSE)
```

```r
saveRDS(so_bm, "data/object/bm_scvi.rds")
```

```r
# Find clusters, then run UMAP, and visualize
so_blood <- FindNeighbors(so_blood, dims=1:10, reduction="latent", assay="SCVI", graph.name=c("SCVI_nn", "SCVI_snn"), return.neighbor=FALSE, verbose=FALSE)
so_blood <- FindClusters(so_blood, algorithm=4, graph.name="SCVI_snn", random.seed=random_seed, verbose=FALSE)
so_blood <- RunUMAP(so_blood, dims=1:10, reduction="latent", umap.method="umap-learn", seed.use=random_seed, verbose=FALSE)
```

```r
saveRDS(so_blood, "data/object/blood_scvi.rds")
```
