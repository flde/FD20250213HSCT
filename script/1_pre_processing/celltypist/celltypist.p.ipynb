```python
adata=sc.read_h5ad('data/object/raw.h5ad')
```

```python
adata = adata[adata.obs['nCount_RNA']>=1000]
adata = adata[adata.obs['pMT_RNA']<=15]
```

```python
sc.pp.normalize_total(adata, target_sum=10000)
sc.pp.log1p(adata)
```

# CellTypist models

```python
models.download_models(force_update=True)
```

# Predict cell types (Immune_All_High.pkl)

```python
predictions=celltypist.annotate(adata, model='Immune_All_High.pkl', majority_voting=False)
```

```python
adata=predictions.to_adata()
adata.obs[['predicted_labels']].to_csv('result/celltypist/immune_all_high.csv')
```
