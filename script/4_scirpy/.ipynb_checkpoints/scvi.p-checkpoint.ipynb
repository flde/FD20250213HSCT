# Import AnnData

```python
adata = sc.read_h5ad('data/object/pp.h5ad')
adata = adata.raw.to_adata()
adata.raw = adata
```

```python
meta = pd.read_csv('data/object/components/meta.csv', index_col=0)
```

```python
tcr = pd.read_csv('result/scirpy/scirpy.csv', index_col=0)
meta = pd.merge(meta, tcr, left_index=True, right_index=True, how='left')
```

```python
genotype = pd.read_csv('result/genotype/genotype_mode_1a_genome1k.csv', index_col='cell_id')[['genotype_class']]
meta = pd.merge(meta, genotype, left_index=True, right_index=True, how='left')
```

```python
adata.obs = meta
```

```python
adata.obs['has_ir'] = adata.obs['has_ir'].map({True:'TCR'})
```

# UMAP

```python
sc.pl.umap(adata, color=['cell_type_leiden', 'has_ir', 'genotype_class', 'hto_demux_class', 'time_point'], frameon=False, ncols=25, wspace=0.5)
```

# Subset for T-cell and ILC leiden cluster

```python
adata = adata[adata.obs['cell_type_leiden'].isin(['TC-1', 'TC-2', 'TC-3', 'TC-4', 'TC-5', 'TC_NK', 'NK', 'ILCs']), ].copy()
adata = adata[adata.obs['genotype_class'].isin(['Host', 'Donor']), ].copy()
```

```python
adata = adata[:, (adata.X>3).sum(axis=0)>=1].copy()
```

```python
sc.pl.umap(adata, color=['cell_type_leiden', 'has_ir', 'genotype_class', 'hto_demux_class', 'time_point'], frameon=False, ncols=25, wspace=0.5)
```

# Register model

```python
scvi.model.SCVI.setup_anndata(
    
    adata,
    categorical_covariate_keys=['patient_id', 'genotype_class'], 
    continuous_covariate_keys=['pMT_RNA']
    
)
```

# Creating and training a model 

```python
model = scvi.model.SCVI(
    
    adata, 
    n_latent=10, 
    n_hidden=128, 
    n_layers=2, 
    gene_likelihood='nb'
    
)
```

```python
model.train()
```

```python
model.save('data/object/scvi/model_tcr/', overwrite=True)
```

# Update anndata object with scvi results 

```python
adata.layers['data'] = model.get_normalized_expression(library_size=10e4)
adata.obsm['latent'] = model.get_latent_representation()
```

# Dimensional reduction 

```python
sc.pp.neighbors(adata, n_neighbors=50, use_rep='latent')
sc.tl.umap(adata, min_dist=1)
```

```python
sc.pl.umap(adata, color=['cell_type_leiden', 'has_ir', 'genotype_class', 'hto_demux_class', 'time_point'], frameon=False, ncols=25, wspace=0.5)
```

# Save scvi output adata

```python
adata.write('data/object/scvi/adata_tcr.h5ad')
```
