```python
# Load color 
import rpy2.robjects as robjects
color_load = robjects.r.source('plotting_global.R')
color = dict()
for i in range(len(color_load[0])):
    color[color_load[0].names[i]] = {key : color_load[0][i].rx2(key)[0] for key in color_load[0][i].names}
```

```python
# Set color function 
def set_color(categories): 
    
    categories = [x for x in categories if x in list(adata.obs.columns)]

    for category in categories: 

        adata.obs[category] = pd.Series(adata.obs[category], dtype='category').cat.remove_unused_categories()
        
        keys = list(color[category].keys())
        keys = [x for x in keys if x in list(adata.obs[category])]

        adata.obs[category] = adata.obs[category].cat.reorder_categories(keys)
        adata.uns[category+'_colors'] = np.array([color[category].get(key) for key in keys], dtype=object)
```

```python
# Seurat object 
qc_h5ad_file = 'data/object/qc.h5ad'

pp_h5ad_file = 'data/object/pp.h5ad'
pp_0_h5ad_file = 'data/object/pp_0.h5ad'
pp_1_h5ad_file = 'data/object/pp_1.h5ad'

pp_rds_file = 'data/object/pp.rds'
pp_0_rds_file = 'data/object/pp_0.rds'
pp_1_rds_file = 'data/object/pp_1.rds'

# Meta data 
annotation_file = 'result/annotation/annotation.csv'
genotype_file = 'result/genotype/genotype.csv'
ir_file = 'result/ir/tcrdist/cell_ir.csv'
```

# Import 

```python
adata = sc.read_h5ad(qc_h5ad_file)
```

```python
annotation = pd.read_csv(annotation_file, index_col=0)
genotype = pd.read_csv(genotype_file, index_col=0)
ir = pd.read_csv(ir_file, index_col=0).drop(columns=["patient_id", "time_point", "hto_demux_class", "genotype_class"])
```

# Meta data 

```python
adata.obs = adata.obs.merge(annotation, left_index=True, right_index=True, how='left')
adata.obs = adata.obs.merge(genotype, left_index=True, right_index=True, how='left')
adata.obs = adata.obs.merge(ir, left_index=True, right_index=True, how='left')
```

# Filter data 

```python
adata = adata[adata.obs.cell_id.isin(annotation.index.to_list())]
adata = adata[adata.obs.celltype_high != '$Artefact$']
```

```python
adata = adata.copy()
```

# Re-name cell types

```python
adata.obs = adata.obs.applymap(lambda x: x.replace('cell', 'cells') if isinstance(x, str) else x)
```

```python
rename_dict = {
    
    "$Fibroblast$": "$Fibroblasts$",
    "$ILC$": "$NK\\ cells$",
    "$Monocyte$": "$Monocytes$",
    "$Macrophage$": "$Macrophages$",
    "$cDC$": "$Dendritic\\ cells$",
    "$LC$": "$Dendritic\\ cells$",
    "$pDC$": "$Dendritic\\ cells$"
}

adata.obs['celltype_high'] = adata.obs['celltype_high'].replace(rename_dict)
```

# SCVI on all samples 

```python
adata = adata[:, (adata.X>=3).sum(axis=0)>=3]
```

```python
adata = adata.copy()
```

```python
cache_scvi = True
```

```python
if not cache_scvi: 
    
    scvi.model.SCVI.setup_anndata(
    
        adata,
        batch_key='sample_name', 
        continuous_covariate_keys=['S_ucell', 'G2M_ucell', 'pct_counts_MT'], 
        labels_key='celltype_low'
        
    )
    
    model = scvi.model.SCVI(
    
        adata, 
        n_latent=30, 
        n_hidden=128, 
        n_layers=2, 
        gene_likelihood='nb'

    )
    
    max_epochs = int(np.min([round((20000 / adata.n_obs) * 400), 400]))
    
    model.train(max_epochs=max_epochs, check_val_every_n_epoch=1)
    
    model.save('data/object/scvi/pp/model/', overwrite=True)
    
else: 
    
    model = scvi.model.SCVI.load('data/object/scvi/pp/model/', adata=adata)
```

```python
cache_dim = True
```

```python
if not cache_dim: 
    
    adata.obsm['latent'] = model.get_latent_representation()
    
    sc.pp.neighbors(adata, n_neighbors=30, use_rep='latent')
    sc.tl.umap(adata, min_dist=1)
    
    adata.write_h5ad('data/object/scvi/pp/adata.h5ad')
    
else: 
    
    adata = sc.read_h5ad('data/object/scvi/pp/adata.h5ad')
```

```python
# Update IR 
adata.obs = adata.obs.drop(columns=['receptor_type', 'receptor_subtype', 'chain_pairing', 'clone_id', 'clone_id_size', 'clonal_expansion', 'species_vdjdb', 'antigen_vdjdb', 'antigen_iedb', 'species_iedb'])
adata.obs = adata.obs.merge(ir.drop(columns=['patient_id', 'time_point', 'hto_demux_class', 'genotype_class']), left_index=True, right_index=True, how="left")
```

```python
sc.pl.umap(adata, color=['celltype_low', 'celltype_high', 'celltype_domain', 'patient_id', 'hto_demux_class', 'time_point', 'msCC_label', 'genotype_class', 'clone_cell_n', 'graph_clone_cell_n'], size=20, ncols=5, wspace=0.1, legend_loc='on data')
```

```python
adata = adata.raw.to_adata()
adata.raw = adata.copy()
```

```python
adata.write_h5ad(pp_h5ad_file)
```

## Save all sample RDS

```python
adata_r.obs['celltype_low'] = adata_r.obs['celltype_low'].cat.reorder_categories(celltype_low.values())
adata_r.obs['celltype_high'] = adata_r.obs['celltype_high'].cat.reorder_categories(celltype_high.values())
adata_r.obs['celltype_domain'] = adata_r.obs['celltype_domain'].cat.reorder_categories(celltype_domain.values())
```

```python
cnt = adata_r.X.T.todense()
cell_id = adata_r.obs_names
gene_id = adata_r.var_names
meta = adata_r.obs
umap = adata_r.obsm['X_umap']
```

```python
%%R -i cnt -i cell_id -i gene_id -i meta -i umap -i pp_rds_file

colnames(cnt) <- cell_id
rownames(cnt) <- gene_id

so <- Seurat::CreateSeuratObject(cnt, meta.data=meta, assay="RNA")

colnames(umap) <- c("UMAP_1", "UMAP_2")
rownames(umap) <- cell_id
so[["umap"]] <- Seurat::CreateDimReducObject(embeddings=as.matrix(umap), key="UMAP_", assay="RNA")

saveRDS(so, pp_rds_file)
```
