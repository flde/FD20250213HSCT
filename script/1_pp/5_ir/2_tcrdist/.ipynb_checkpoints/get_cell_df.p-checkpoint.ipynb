# Loading and preparing data 

```python
import scanpy as sc

import pandas as pd
import numpy as np

import os
import sys
import warnings
```

## Root directory

```python
os.chdir('/research/peer/fdeckert/FD20250213HSCT')
```

# Import AnnData

```python
adata=sc.read_h5ad('data/object/qc.h5ad')
```

## Sample path of 10x TCR data 

```python
adata.obs['sample_path_tcr']=np.where(adata.obs['sample_path'].str.contains('S_'), adata.obs['sample_path'].astype(str).replace('_transcriptome', '_VDJ', regex=True), adata.obs['sample_path'].astype(str).replace('_transcriptome', '_abTCR_VDJ', regex=True))
```

## Subset for annotated cells 

```python
annotation_file = 'result/annotation/annotation.csv'
annotation = pd.read_csv(annotation_file, index_col=0)
```

```python
adata.obs = adata.obs.merge(annotation, left_index=True, right_index=True, how='left')
```

```python
adata = adata[adata.obs.celltype_high!='$Artefact$']
adata = adata[~adata.obs.celltype_high.isna()]
```

## Add genotype 

```python
genotype_file = 'result/genotype/genotype.csv'
genotype = pd.read_csv(genotype_file, index_col=0)
```

```python
adata.obs = adata.obs.merge(genotype, left_index=True, right_index=True, how='left')
```

```python
adata = adata[adata.obs.genotype_class.isin(['Donor', 'Host'])]
```

# Import TCR 

```python
cell_df=[]
for sample_path_tcr in adata.obs['sample_path_tcr'].unique(): 
    
    print(sample_path_tcr)
    
    # Import TCR 10x contig and airr data 
    contig = pd.read_csv(sample_path_tcr+'/filtered_contig_annotations.csv')
    airr = pd.read_csv(sample_path_tcr+'/airr_rearrangement.tsv', sep='\t')
    
    # keep productive, high-confidence TCR chains
    c = contig[(contig['productive'] == True) & (contig['high_confidence'] == True) & (contig['chain'].isin(['TRA','TRB']))].copy()
    
    # optional: add AIRR counts (if present) to rank chains per cell
    a = airr[['sequence_id','duplicate_count','consensus_count']].copy()
    c = c.merge(a, left_on='contig_id', right_on='sequence_id', how='left')

    rank_cols = ['barcode', 'duplicate_count', 'consensus_count', 'umis', 'reads']
    for col in ['duplicate_count','consensus_count','umis','reads']:
        if col not in c.columns: c[col] = 0

    alpha = (c[c['chain']=='TRA'].sort_values(rank_cols, ascending=[True, False, False, False, False]).drop_duplicates('barcode').rename(columns={'cdr3':'cdr3_a_aa', 'cdr3_nt':'cdr3_a_nucseq', 'v_gene':'v_a_gene', 'j_gene':'j_a_gene'})[['barcode', 'cdr3_a_aa', 'cdr3_a_nucseq', 'v_a_gene', 'j_a_gene']])
    beta = (c[c['chain']=='TRB'].sort_values(rank_cols, ascending=[True, False, False, False, False]).drop_duplicates('barcode').rename(columns={'cdr3':'cdr3_b_aa', 'cdr3_nt':'cdr3_b_nucseq', 'v_gene':'v_b_gene', 'j_gene':'j_b_gene'})[['barcode', 'cdr3_b_aa', 'cdr3_b_nucseq', 'v_b_gene', 'j_b_gene']])

    # Create TCRdist df
    df = alpha.merge(beta, on='barcode', how='inner')

    # Get GEX barcodes index of the merged AnnData object
    obs_names=adata[adata.obs['sample_path_tcr']==sample_path_tcr].obs_names
    barcode_idx='_'+obs_names[0].split('_')[1]
    print(barcode_idx)

    # Append cell df barcodes with index and filter 
    df.barcode=df.barcode+barcode_idx
    df = df[df.barcode.isin(obs_names)]

    # Add obs information 
    obs = adata[adata.obs['sample_path_tcr']==sample_path_tcr].obs.copy()
    obs['barcode'] = obs['cell_id']
    obs['subject'] = obs['patient_id']

    df = df.merge(obs[['barcode', 'subject', 'time_point', 'hto_demux_class', 'genotype_class']], on='barcode', how='inner')

    # Fix gene names 
    def add_star01(g):
        if pd.isna(g): 
            return g
        g = str(g).strip()
        if g == '' or g == 'None':
            return pd.NA
        if '*' in g:
            return g
        return g + '*01'

    for col in ['v_a_gene','j_a_gene','v_b_gene','j_b_gene']:
        df[col] = df[col].map(add_star01)
    
    # if any CDR3 fields are empty strings, turn them into NA (prevents “captured=0”)
    for col in ['cdr3_a_aa','cdr3_b_aa','cdr3_a_nucseq','cdr3_b_nucseq']:
        if col in df.columns:
            df[col] = df[col].replace(r'^\s*$', pd.NA, regex=True)
    
    # keep only fully paired rows (or relax if doing beta-only)
    df = df.dropna(subset=['cdr3_a_aa','v_a_gene','j_a_gene','cdr3_b_aa','v_b_gene','j_b_gene'])

    # Output to dict
    cell_df.append(df)
```

```python
cell_df = pd.concat(cell_df, ignore_index=True)
```

```python
cell_df.to_csv('result/ir/tcrdist/cell_df.csv')
```
