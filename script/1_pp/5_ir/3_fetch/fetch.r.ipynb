# Parameter 

```r
cell_df <- read.csv("result/ir/tcrdist/cell_df.csv", row.names=1) %>% tibble::rowid_to_column("row_id") %>% tibble::column_to_rownames("row_id") %>% 
    dplyr::group_by(clone_id) %>% dplyr::mutate(clone_cell_n=n()) %>% dplyr::ungroup() %>% 
    dplyr::mutate(cdr3_a_aa_length=nchar(cdr3_a_aa), cdr3_b_aa_length=nchar(cdr3_b_aa)) 
```

```r
clone_graph <- read.csv("result/ir/tcrdist/graph_clone_df.csv", row.names=1) %>% tibble::rowid_to_column("row_id") %>% tibble::column_to_rownames("row_id")
```

# Combine results 

```r
cell_ir <- cell_df
```

```r
cell_ir <- dplyr::left_join(cell_ir, clone_graph %>% dplyr::select(v_a_gene, j_a_gene, cdr3_a_aa, v_b_gene, j_b_gene, cdr3_b_aa, graph_clone_id, graph_clone_n), by=join_by(v_a_gene, j_a_gene, cdr3_a_aa, v_b_gene, j_b_gene, cdr3_b_aa)) %>% 
    dplyr::group_by(graph_clone_id) %>% dplyr::mutate(graph_clone_cell_n=n()) 
```

# Save results

```r
# By cell id results 
cell_ir %>% tibble::column_to_rownames("barcode") %>% write.csv("result/ir/tcrdist/cell_ir.csv")
```
