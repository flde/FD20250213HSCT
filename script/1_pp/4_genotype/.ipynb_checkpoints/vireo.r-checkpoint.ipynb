# Function to load and merge vireo data with Seurat object by cell_id

```r
vireo_load <- function(so, dir_suffix, out_file) {
    
    # Allocate result Seurat list 
    so_vireo <- list()

    # Merge vireo results with Seurat objects by cell_id
    for (i in unique(so$patient_id)) {

        # Load vireo data 
        vireo_dir=paste0("/research/peer/fdeckert/FD20220623HSCT/data/object/vireo/", i, dir_suffix)
        vireo_df <- fread(paste0(vireo_dir, "/donor_ids.tsv"))

        # Convert cell column to cell_id and sample_name
        vireo_df$cell_id <- sapply(strsplit(vireo_df$cell, "Baseline|Tx|D14|D100|GVHD"), "[", 2)
        vireo_df$patient_id <- i
        
        vireo_df$time_point <- sapply(vireo_df$cell, function(x) c("Baseline", "Tx", "D14", "D100", "GVHD")[sapply(c("_Baseline", "_Tx", "_D14", "_D100", "_GVHD"), grepl, x)])           
        vireo_df$sample_name <- paste0(vireo_df$patient_id, "_", vireo_df$time_point)
                                      
        
        # Subset Seurat object by patient_id and split by sample name 
        so_patient_id <- subset(so, subset=patient_id==i)
        so_patient_id <- SplitObject(so_patient_id, split.by="sample_name")                          
        
        # Add vireo results to Seurat object 
        for (i in names(so_patient_id)){
            
            so_sample <- so_patient_id[[i]]
            
            so_sample$cell_id <- colnames(so_sample)
            so_sample$cell_id_ident <- sapply(strsplit(colnames(so_sample), "_"), "[", 1)
            
            vireo_df_sample_name <- vireo_df[vireo_df$sample_name==i, ]
            vireo_df_sample_name <- data.frame(vireo_df_sample_name)
            
            rownames(vireo_df_sample_name) <- paste0(vireo_df_sample_name$cell_id, "_", strsplit(colnames(so_sample)[1], "_")[[1]][2])
            vireo_df_sample_name <- dplyr::select(vireo_df_sample_name, -cell_id, -patient_id, -time_point, -sample_name)

            so_sample <- AddMetaData(so_sample, vireo_df_sample_name)
            so_patient_id[[i]] <- so_sample

        }
                                      
        if(length(so_patient_id) > 1) {
            
            so_patient_id <- merge(so_patient_id[[1]], so_patient_id[2:length(so_patient_id)])
        
        } else {
            
            so_patient_id <- so_patient_id[[1]]
        
        }
                                      
        so_vireo[[i]] <- so_patient_id

        rm(so_patient_id)
        rm(so_sample)
        rm(vireo_df)
        rm(vireo_df_sample_name)
                                      
    }

    so <- merge(so_vireo[[1]], so_vireo[2:length(so_vireo)])
    vireo_meta <- dplyr::select(so@meta.data, cell, donor_id, prob_max, prob_doublet, n_vars, best_singlet, best_doublet, doublet_logLikRatio)                     
    write.csv(vireo_meta, out_file)
                                      
    return(vireo_meta)
    
}
```

```r
if(compute) {vireo <- vireo_load(so, "_vireo", "result/vireo/vireo.csv")} else {vireo <- read.csv("result/vireo/vireo.csv", row.names=1)}
```
