# Parameter settings

```r
# Files 
so_rds_file <- "data/object/qc.rds"

# Meta data 
meta <- read.csv("data/object/components/meta.csv", row.names=1)

# UMAP 
umap <- read.csv("data/object/components/umap.csv", row.names=1)
colnames(umap) <- c("UMAP_1", "UMAP_2")
```

# Import Seurat object

```r
so <- readRDS(so_rds_file)
```

```r
# Set meta data
so <- AddMetaData(so, meta)
```

```r
# Set UMAP
so[["umap"]] <- CreateDimReducObject(embeddings=as.matrix(umap), key="UMAP_")
```

```r
# Subset 
so <- subset(so, subset=sample_name %in% c("T0125_1", "T0125_2"))
```

# Import msigdbr

```r
gs_mm <- msigdbr(species="Mus musculus", category="C2", subcategory="WIKIPATHWAYS")
gs_mm <- split(x=gs_mm$gene_symbol, f=gs_mm$gs_name)
```

## Check cell types if all cell types are present in the DEA subset

```r
# DP is missing 
cell_type_leiden <- cell_type_leiden[cell_type_leiden$cell_type_leiden_fine %in% so$cell_type_leiden_fine, ]
```

## Check if all cell types have both groups present for DEA

```r
table <- dplyr::group_by(so@meta.data, cell_type_leiden_fine, genotype) %>% 
    dplyr::summarise(n=n()) %>% 
    reshape2::dcast(., formula=genotype~cell_type_leiden_fine) %>% 
    column_to_rownames(var="genotype") %>% 
    replace(is.na(.), 0)

knitr::kable(table, "html") %>% as.character() %>% IRdisplay::display_html()
```

```r
cell_type_leiden <- cell_type_leiden[!cell_type_leiden$cell_type_leiden_fine %in% c("DC", "Endothelial cell (3)", "Granulocyte (2)", "Granulocyte (3)", "pDC"), ]
so <- subset(so, subset=cell_type_leiden_fine %in% cell_type_leiden$cell_type_leiden_fine)
```

# DEA 

```r
dea <- dea_seurat(
    
    so=so, 
    mode="group", 
    file="result/dea/baseline_adar_wt_adar_ko", 
    ident="cell_type_leiden_fine",  
    map=cell_type_leiden, 
    grouping_var="genotype", 
    ident_1="Adarb1KO", 
    ident_2="Ctr", 
    only_pos=FALSE, 
    logfc_threshold=0, 
    min_pct=0, 
    compute=FALSE, 
    test_use="MAST", 
    cnt_min=3, 
    cell_min=3, 
    gene_filter=NULL
    
)
```

```r
options(repr.plot.width=5*6, repr.plot.height=ceiling(length(dea)/5)*6)

vp_treatment <- lapply(seq_along(dea), function(i) vp_dea(dea[[i]], title=dea[[i]]$ident, log2_thold=0, adjpvalue_thold=0.1) + theme_global_set(size_select=1))
wrap_plots(vp_treatment, ncol=5, nrow=ceiling(length(dea)/5)) 
```

# GSEA

```r
dea <- dea[unlist(lapply(dea, function(x) sum(x$p_val_adj<=0.05)>=5))]
```

```r
options(repr.plot.width=2*20, repr.plot.height=ceiling(length(dea)/2)*10)

gsea <- lapply(1:length(dea), function(i) gsea_from_dea(dea=dea[[i]], gene_sets=gs_mm, ident=names(dea)[i]))
gsea_plot <- lapply(gsea, function(x) {gsea_plot(x, pval_thr=0.1, pathway_suffix="WP", color_point=setNames(c(RColorBrewer::brewer.pal(8, "Set1")[1], RColorBrewer::brewer.pal(8, "Set1")[2]), c("Adarb1KO", "Ctr"))) + theme_global_set(size_select=1)})
wrap_plots(gsea_plot, ncol=2)
```
