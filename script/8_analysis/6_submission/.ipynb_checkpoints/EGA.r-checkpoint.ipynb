# Meta data 

```r
meta <- read.csv("meta.csv") %>% 
    # Re-set patient id
    dplyr::left_join(., data.frame(patient_name=c(763, 764, 766, 768, 783, 784, 785, 4, 7, 8), patient_id_new=paste0("Patient_", 1:10)), by=join_by(patient_name)) %>% 
    dplyr::mutate(patient_id=patient_id_new) %>% 
    dplyr::mutate(patient_id=ifelse(is.na(patient_id), "Patient_11", patient_id)) %>% 
    # Re-set sample_name
    dplyr::mutate(sample_name=paste0(patient_id, "_", time_point)) %>%
    # Set GEO path 
    dplyr::mutate(ega_path=paste0("/nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/", tolower(sample_name)))
```

## Transfer GEX BAM

```r
for (i in 1:nrow(meta)){
    
    # Set sample meta data 
    x <- meta[i, , drop=FALSE]
    
    # Specify file paths
    input_bam <- paste0(x$sample_path, "/possorted_genome_bam.bam")
    output_bam <- paste0(x$ega_path, "_possorted_genome_bam.bam")

    # Use samtools to extract header and first 10 entries
    system(paste0("/cm/shared/specific/apps/SAMtools/1.17-GCC-11.3.0/bin/samtools view -H ", input_bam, " > /nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/header.sam"))
    system(paste0("/cm/shared/specific/apps/SAMtools/1.17-GCC-11.3.0/bin/samtools view ", input_bam, " | head -n 10 > /nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/record.sam"))

    # Combine header and first 10 entries
    system(paste0("cat /nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/header.sam /nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/record.sam > ", sub("\\.bam$", ".sam", output_bam)))

    # Convert SAM to BAM
    system(paste0("/cm/shared/specific/apps/SAMtools/1.17-GCC-11.3.0/bin/samtools view -S -b ", sub("\\.bam$", ".sam", output_bam), " > ", output_bam))

    # Clean up intermediate files
    file.remove("/nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/header.sam", "/nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/record.sam", sub("\\.bam$", ".sam", output_bam))

}
```

## Transfer TCR BAM

```r
for (i in 1:nrow(meta)){
    
    # Set sample meta data 
    x <- meta[i, , drop=FALSE]
    
    # Get TCR directory 
    if(grepl("CoR_", x$sample_path)) {x$tcr_path <- gsub("_transcriptome", "_abTCR_VDJ", x$sample_path)}
    if(grepl("GvD_", x$sample_path)) {x$tcr_path <- gsub("_transcriptome", "_abTCR_VDJ", x$sample_path)}
    if(grepl("S_", x$sample_path)) {x$tcr_path <- gsub("_transcriptome", "_VDJ", x$sample_path)}
    if(grepl("HSCT_", x$sample_path)) {x$tcr_path <- gsub("_transcriptome", "_abTCR_VDJ", x$sample_path)}
    
    # Specify file paths
    input_bam <- paste0(x$tcr_path, "/all_contig.bam")
    output_bam <- paste0(x$ega_path, "_all_contig.bam")

    # Use samtools to extract header and first 10 entries
    system(paste0("/cm/shared/specific/apps/SAMtools/1.17-GCC-11.3.0/bin/samtools view -H ", input_bam, " > /nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/header.sam"))
    system(paste0("/cm/shared/specific/apps/SAMtools/1.17-GCC-11.3.0/bin/samtools view ", input_bam, " | head -n 10 > /nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/record.sam"))

    # Combine header and first 10 entries
    system(paste0("cat /nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/header.sam /nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/record.sam > ", sub("\\.bam$", ".sam", output_bam)))

    # Convert SAM to BAM
    system(paste0("/cm/shared/specific/apps/SAMtools/1.17-GCC-11.3.0/bin/samtools view -S -b ", sub("\\.bam$", ".sam", output_bam), " > ", output_bam))

    # Clean up intermediate files
    file.remove("/nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/header.sam", "/nobackup/peer/fdeckert/FD20220623HSCT/data/EGA/scrnaseq/record.sam", sub("\\.bam$", ".sam", output_bam))


}
```
